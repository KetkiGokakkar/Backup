Business Scenario:
****************
You are a Python Developer working for an online retail store. The store wants to build a basic shopping cart system to help manage products, customers, and their orders.

The system should be able to:
************************
	Store available products with their prices (Product Catalog)
	Allow customers to add or remove items from their cart
	Ensure unique customer accounts (no duplicates allowed)
	Keep each customer’s details fixed once registered (ID, Name, Email)
	At checkout, calculate the total bill for the customer’s cart

Sample Dataset:
**************
1. Product Catalog (Dictionary)

```python
products = {
    1: ("Laptop    ", 1200.00),
    2: ("Smartphone", 600.00),
    3: ("Headphones", 50.00),
    4: ("Smartwatch", 200.00)
}
```

2. Customers (Set to ensure uniqueness)

```python
customers = {
    (101, "Dinesh Karthik", "dk@example.com"),
    (102, "Ajay Aravind", "aa@example.com")
}
```

3. Example Shopping Cart (List of Tuples)

```python
cart = [
    (1, 2),   # 2 Laptops
    (3, 1)    # 1 Headphones
]
```

Tasks to be Completed:
*******************

1. Display Available Products -- Show all products with IDs, names, and prices.

```python
print("Product ID","Name","Price",sep='         ')
for i ,j in products.items():
    print(i,j[0],j[1],sep='              ')
```

2. Add an Item to Car -- A customer can select a product by ID and specify the quantity to add.

```python
cart.append((2,2))
print(cart)
```

3. Remove an Item from Cart -- A customer can remove a product from the cart by specifying its ID.

```python
for i in cart :
    if 1 == i[0] :
        cart.remove(i)
        
```

4. Prevent Duplicate Customers-- Ensure that when registering new customers, IDs or emails are unique.

```python
def insert_cus(id , name,email):
    for i in customers:
        if id ==i[0] or email==i[2]:
            print("Customer Already Exists")
            break 
    else :
        customers.add((id,name,email))

insert_cus(101,'abc','abc@gmail.com')
print(customers)
```

5. Checkout and Calculate Total Bill -- For each product in the cart, multiply quantity × price and calculate the total bill.

```python
tot_bill = 0 
for i in cart:
    tot_bill += i[1]* products[i[0]][1]
print(tot_bill)
```

# Q1: Business Scenario: A retail company wants to implement a dynamic discount system for different types of customers.

    Regular customers get 10% discount if their purchase amount exceeds ₹500.
    Premium customers get 15% discount if their purchase amount exceeds ₹1000.
    Gold customers get 20% discount if their purchase amount exceeds ₹2000.
    Any other customer type does not receive any discount.

# Write a Python function calculate_discounted_price(customer_type, purchase_amount) that:

    Uses a dictionary to store discount thresholds and percentages for each customer type.
    Looks up the discount based on the customer type. ("regular" or "premium") and their purchase amount.
    Applies the correct discount based on the rules above.
    Applies the discount only if the purchase amount crosses the threshold.
    Returns the final price after applying the discount.
    
USE THE BELOW FOR DISCOUNT CONFIGRUATION IN YOUR FUNCTION:
***********************************************************
# Discount configuration using a dictionary
 

```python
discounts = {
        "regular": {"threshold": 500, "discount": 0.10},
        "premium": {"threshold": 1000, "discount": 0.15},
        "gold": {"threshold": 2000, "discount": 0.20}
}
```

```python
def calculate_discounted_price(customer_type, purchase_amount):
    if purchase_amount>  discounts[customer_type]["threshold"]:
        discount= purchase_amount* discounts[customer_type]["discount"]  
    else:
        discount= 0
    return purchase_amount-discount
```

```python
print(calculate_discounted_price("regular",4000))
```
