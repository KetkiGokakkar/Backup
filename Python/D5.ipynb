```python
class Employee:
    def __init__(self , emp_id  ,name , salary):
        self.emp_id = emp_id 
        self.name = name 
        self.salary = salary 

    def display_details(self):
        print(f"ID: {self.emp_id}")
        print(f"Name: {self.name}")
        print(f"Salary : {self.salary}")

    def apply_bonus(self,percentage):
        bonus= self.salary * (percentage/100)
        self.salary +=bonus 
        print(f"Bonus of {percentage}% applied. New salary: ₹{self.salary}")

emp1= Employee(101, "Arun",50000)

emp1.display_details()
emp1.apply_bonus(20)
```

```python
class Car:

    def __init__(self , make , model , year, milage , fuel_level ):
        self.make=make 
        self.model = model 
        self.year = year
        self.milage=milage
        self.fuel_level = fuel_level 
        self.maintain =[]
    def drive(self,miles):
        self.milage += miles
        self.fuel_level -= miles*0.05

    def refuel(self,gallons):
        self.fuel_level+=gallons 
    def add_maint(self,date,note):
        self.maintain.append({"Date":date,"Note":note})

    def get_info(self) :
        print(self.fuel_level)

    def get_maintain_rec(self):
        print(self.maintain)
```

```python
car1 = Car("Toyoto","8dg38",2002,12000,50)
car1.drive(20)
car1.get_info()
```
