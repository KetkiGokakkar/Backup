A Pandas Series is a one-dimensional labeled array that can hold any data type (e.g., integers, strings, floats, etc.).

# Key Points:

    Labeled: Each element in the Series has a label (index), which can be customized.
    Homogeneous: The data in a Series are typically of the same type (though it can hold different types).

# Common Ways to Create a Series:

# Series: One-Dimensional Labeled Arrays

A Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floats, etc.). 
It is similar to a list, but it has labels (called index) for each element.

# 1. From a list:

```python
import pandas as pd

data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)
print(type(series))
```

# Here, the index is automatically generated (0, 1, 2, 3, 4).

# 2. From a NumPy array:

```python
import numpy as np
import pandas as pd

np_array = np.array([5.5, 10.5, 15.5, 20.5])
series = pd.Series(np_array)
print(series)
print(type(np_array))
print(type(series))
```

```python
import numpy as np
import pandas as pd

series_from_array = pd.Series(np.array([1.5, 2.5, 3.5]))
print(series_from_array)
```

# 3. From a dictionary (keys become the index, values become the data):

```python
import numpy as np
import pandas as pd

data_dict = {'a': 1, 'b': 2, 'c': 3}
series = pd.Series(data_dict)
print(series)
print(type(data_dict))
print(type(series))
```

# Attributes of a Series:
    index: Returns the index (labels) of the Series.
    values: Returns the data of the Series as a NumPy array.
    dtype: Returns the data type of the elements in the Series.

```python
print(series.index)   # Index: ['a', 'b', 'c']
print(series.values)  # Values: [1, 2, 3]
print(series.dtype)   # DataType: int64
```

# Common Methods of Series:  
    head(): Returns the first n elements.
    tail(): Returns the last n elements.
    describe(): Generates descriptive statistics of the Series.
    sum(): Returns the sum of the elements.
    mean(): Returns the mean of the elements.
    map(): Apply a function to each element.

```python
# Example:
```

```python
import pandas as pd
series = pd.Series([10, 20, 30, 40, 50])

print(series.head(3))     # First 3 elements
print(series.tail(2))     # Last 2 elements
print(series.describe())  # Descriptive stats
print(series.sum())       # Sum of elements
print(series.mean())      # Mean of elements

squared = series.map(lambda x: x**2)
print(squared)
```

# Indexing and Slicing:
    
You can access elements by their index:

```python
# With Index Label:
import pandas as pd
series = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])

print(series)
print(series['c'])  # Output: 30
```

```python
# Without Index Label:
import pandas as pd
series = pd.Series([10, 20, 30, 40, 50])

print(series)
print(series[2])  # Output: 30
```

```python
import pandas as pd
series = pd.Series([10, 20, 30, 40, 50], index=[1,2,3,4,5])

print(series)
print(series[2:4])  # Output: 30,40
```

# You can also slice the Series:

```python
import pandas as pd
series = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])

print(series['b':'d'])  # Slices from 'b' to 'd'
```

# Example of Series Creation and Basic Operations:

```python
import pandas as pd

# Creating a Series from a list
data = [100, 200, 300, 400, 500]
index = ['a', 'b', 'c', 'd', 'e']
series = pd.Series(data, index=index)

print("Series:")
print(series)

# Basic operations
print("\nSum of elements:", series.sum())
print("Mean of elements:", series.mean())
print("First 3 elements:")
print(series.head(3))
```

# In this example:

The series is a Pandas Series with custom indices ['a', 'b', 'c', 'd', 'e'].
The operations sum() and mean() compute the sum and the average of the elements.
The head(3) method returns the first three elements.
