# 2. Create Basic Plots (Line Plots, Scatter Plots) using pyplot

# The pyplot module of Matplotlib provides a MATLAB-like interface for creating plots with just a few lines of code. 
# Below are examples of how to create basic plots.

# Example 1: Line Plot:
        
A line plot is one of the simplest and most common types of plots. It is used to visualize the 
relationship between two continuous variables.

```python
import matplotlib.pyplot as plt  # matplotlib.pyplot is used for plotting

# Data
x = [0, 1, 2, 3, 4, 5]           # x represents the independent variable (X-axis values).
y = [0, 1, 4, 9, 16, 25]         # y represents the dependent variable (Y-axis values). (Each y value is the square of the corresponding x value.)

# Create line plot
plt.plot(x, y)    # Plots the (x, y) data points as a continuous line. By default, Matplotlib connects the points with a solid blue line.

# Add labels and title
plt.xlabel('X-axis')           # Labels the X-axis (X-axis) 
plt.ylabel('Y-axis')           # Labels the Y-axis (Y-axis).
plt.title('Simple Line Plot')  # Adds a title (Simple Line Plot).

# Show the plot
plt.show()  # Renders the plot./ Displays the generated plot.
```

# Explanation:

plt.plot(x, y): Creates the line plot with x and y data.
plt.xlabel() and plt.ylabel(): Set the labels for the x and y axes.
plt.title(): Sets the title for the plot.
plt.show(): Displays the plot.

```python
# Customized Line Chart with additional Properties like - Marker , linestyle , color, label

import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 50]

# Create a line plot
plt.plot(x, y, marker='o', linestyle='-', color='b', label='Sample Data')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Line Plot')
plt.legend()

# Show the plot
plt.show()
```

# Example 2: Scatter Plot:
    
A scatter plot is used to visualize the relationship between two continuous variables by displaying points on a Cartesian plane.

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create scatter plot
plt.scatter(x, y)

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Scatter Plot')

# Show the plot
plt.show()
```

# Explanation:

plt.scatter(x, y): Creates a scatter plot with x and y data.
Labels and title are added in the same way as in the line plot.

# 3. The Anatomy of a Matplotlib Figure:
    
A Matplotlib plot consists of several key components. Understanding these components 
helps you customize and improve the quality of your visualizations.

# Main Components of a Plot:
    
    Figure: The overall container for the plot. 
        -- Axes: The area where the data is plotted. A figure can have one or more axes.
        -- Labels: Used for the x-axis and y-axis to provide context.
        -- Title: The title at the top of the plot to describe the data.
    Ticks: The markings on the axes that indicate the data values.
    Legend: If multiple datasets are plotted, a legend can be added for clarity.

# Example 3: Multiple Axes and Customization

```python
import matplotlib.pyplot as plt

# Data
x = [0, 1, 2, 3, 4, 5]
y1 = [0, 1, 4, 9, 16, 25]
y2 = [0, 2, 4, 6, 8, 10]

# Create a figure and a set of subplots (1 row, 2 columns)
fig, (ax1, ax2) = plt.subplots(1, 2)

# Plot on the first axes (ax1)
ax1.plot(x, y1, label='y = x^2', color='r')
ax1.set_xlabel('X-axis')
ax1.set_ylabel('Y-axis')
ax1.set_title('Plot of x^2')
ax1.legend()

# Plot on the second axes (ax2)
ax2.scatter(x, y2, label='y = 2x', color='b')
ax2.set_xlabel('X-axis')
ax2.set_ylabel('Y-axis')
ax2.set_title('Scatter Plot of 2x')
ax2.legend()

# # Adjust layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

# Explanation:

plt.subplots(1, 2): Creates a figure with 1 row and 2 columns of axes.
ax1 and ax2 are individual axes where different plots are drawn.
ax.set_xlabel(), ax.set_ylabel(), and ax.set_title(): Customizes the labels and titles for each subplot.
ax.legend(): Adds a legend for each subplot.
plt.tight_layout(): Adjusts the spacing between subplots to avoid overlap.

# Summary of Key Components:
    
    Figure: The entire plot or canvas.
    Axes: The area where data points or lines are plotted.
    Labels: Provide context for the axes (X and Y).
    Title: Describes what the plot is about.
    Legend: Useful when plotting multiple datasets for clarification.

# 1. Customizing Colors, Markers, and Line Styles:
    
# Example 1: Customizing Line Plot:
        
You can adjust the color, marker style, and line style to make your plots more informative or visually appealing.

```python
import matplotlib.pyplot as plt

# Data
x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]

# Create a line plot with customizations
plt.plot(x, y, color='green', linestyle='--', marker='o', markersize=8, label='y = x^2')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Customized Line Plot')

# Add a legend
plt.legend()

# Show the plot
plt.show()
```

# Customization Options:

    color='green': Changes the color of the line.
    linestyle='--': Sets the line style to dashed.
    marker='o': Adds circle markers at data points.
    markersize=8: Adjusts the size of the markers.
    label='y = x^2': Adds a label for the legend.

# Example 2: Customizing Scatter Plot

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create a scatter plot with customizations
plt.scatter(x, y, color='red', marker='^', s=100, label='y = 2x')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Customized Scatter Plot')

# Add a legend
plt.legend()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create a scatter plot with customization
plt.scatter(x, y, color='purple', marker='s', s=150, label='y = 2x')

# Set titles and labels with customized fonts
plt.title('Advanced Plot Customization', fontsize=16, fontweight='bold', family='serif')
plt.xlabel('X-axis', fontsize=12)
plt.ylabel('Y-axis', fontsize=12)

# Customize grid
plt.grid(True, linestyle=':', color='gray')

# Set axis limits
plt.xlim(0, 6)
plt.ylim(0, 12)

# Add a legend
plt.legend()

# Show the plot
plt.show()
```

# 5. Box Plot:
   
A box plot (or box-and-whisker plot) is used to display the distribution of a dataset, showing the median, quartiles, and outliers.

    The median (middle value),
    Quartiles (which divide the data into four equal parts),
    Outliers (data points that fall outside the typical range).

# Example: Box Plot

# Key Elements of a Box Plot:

    Box: Represents the interquartile range (IQR), which is the middle 50% of the data (between the first quartile (Q1) and the third quartile (Q3)).
    Whiskers: These lines extend from the box to the smallest and largest values within 1.5 * IQR of the first and third quartiles, respectively.
    Median: A line inside the box that indicates the middle value of the data.
    Outliers: Data points that fall outside the whiskers (often shown as individual points).

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data
data = np.random.randn(100)

# Create a box plot
plt.boxplot(data, vert=False, patch_artist=True, boxprops=dict(facecolor='lightblue', color='blue'))

# Add title
plt.title('Box Plot Example')

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some sample data
np.random.seed(10)
data = [np.random.normal(0, std, 100) for std in range(1, 4)]  # Three sets of data with different standard deviations

# Create a box plot
plt.boxplot(data, vert=True, patch_artist=True, tick_labels=['Group 1', 'Group 2', 'Group 3'])  # vert=False Boxes are placed horizontal
#The 'labels' parameter of boxplot() has been renamed 'tick_labels' since Matplotlib 3.9;

# Add title and labels
plt.title("Box Plot Example")
plt.xlabel("Group")
plt.ylabel("Value")

# Show the plot
plt.show()
```

# Explanation of the Code:

Generate Data:

np.random.seed(10) ensures that the random data generated is reproducible.
data = [np.random.normal(0, std, 100) for std in range(1, 4)] generates three groups of data. Each group follows a normal distribution with a mean of 0 and increasing standard deviations (1, 2, and 3).
Create the Box Plot:

plt.boxplot(data) creates the box plot.
vert=True means the boxes are oriented vertically.
patch_artist=True makes the boxes filled with color.
labels=['Group 1', 'Group 2', 'Group 3'] assigns labels to each group of data.
Add Title and Labels:

plt.title("Box Plot Example") adds a title to the plot.
plt.xlabel("Group") and plt.ylabel("Value") label the x and y axes, respectively.
Show the Plot:

plt.show() displays the box plot.

# What the Box Plot Shows:

# Group 1 (with standard deviation of 1):
Has a tighter spread around the median.
Fewer outliers.

# Group 2 (with standard deviation of 2):
Shows a wider spread around the median.
The interquartile range (IQR) is larger than in Group 1.

# Group 3 (with standard deviation of 3):
Displays an even wider spread.
There may be more noticeable outliers due to the larger variability in the data.

# Interpretation:
    Median: The line inside the box indicates the median value of each group. In this example, all groups have a median near 0.
    IQR: The boxes represent the middle 50% of the data. Group 3 has a much wider IQR than Group 1, showing greater data spread.
    Whiskers: The whiskers extend to the smallest and largest values within 1.5 * IQR. Data points outside of this range are considered outliers.
    Outliers: Any data points outside the whiskers are shown as individual dots. For example, in Group 3, some extreme values are shown as outliers.

# Use Case of Box Plots:

Box plots are useful when comparing the distributions of multiple datasets:

Comparing Performance: For example, in machine learning, box plots can be used to compare the performance of different algorithms.
Exploring Variability: They are excellent for understanding the spread and variability of data, especially when working with multiple categories or groups.
Identifying Outliers: Box plots help in identifying outliers, which could either represent important variations in the data or indicate errors or anomalies.

# Conclusion:
Box plots provide a concise summary of the distribution of data and are an excellent tool for comparing multiple datasets. They can reveal important insights about data spread, central tendency, and outliers, making them a vital part of exploratory data analysis (EDA).

# 6. Pie Chart
A pie chart is used to show the proportions of a whole. It is typically used for categorical data with a limited number of categories.

# Example: Pie Chart

```python
import matplotlib.pyplot as plt

# Data
sizes = [40, 30, 20, 10]
labels = ['A', 'B', 'C', 'D']
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen']

# Create a pie chart
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)

# Add title
plt.title('Pie Chart Example')

# Show the plot
plt.show()
```

# 7. Area Plot:

An area plot is similar to a line plot but fills the area under the line. It is useful for visualizing cumulative data or the contribution of each part to the whole.

# Example: Area Plot

```python
import matplotlib.pyplot as plt    # matplotlib.pyplot is used for plotting.
import numpy as np                 # numpy is used for numerical computations, particularly for generating x value

# Data
x = np.linspace(0, 5, 100) # Creates an array of 100 evenly spaced values between 0 and 5.
y1 = np.sin(x)  # Computes the sine values for x
y2 = np.cos(x)  # Computes the cosine values for x

# Create an area plot
plt.fill_between(x, y1, color='lightblue', alpha=0.5, label='sin(x)')
plt.fill_between(x, y2, color='lightgreen', alpha=0.5, label='cos(x)')  # The alpha=0.5 parameter controls transparency (50% opacity).

# Add title and labels
plt.title('Area Plot Example')  # Adds a title (Area Plot Example).
plt.xlabel('X-axis')            # Labels the X-axis (X-axis) and Y-axis (Y-axis)
plt.ylabel('Y-axis')

# Display legend
plt.legend()  # Displays a legend to identify the colored areas

# Show the plot
plt.show()  # Renders the plot.
```
