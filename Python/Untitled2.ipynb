````python
class EmployeeRecordManager:
    def create_employee_record(self,name, department, salary):
        return {
            "name": name,
            "department": department,
            "salary": salary
        }
    
    def get_department(Self,record):
        return record.get("department")
    
    def update_salary(Self,record, new_salary):
        record["salary"] = new_salary
        return record
    
    def add_attribute(self, record, key, value):
        record[key] = value
        return record
````

````python
manager = EmployeeRecordManager()
employee = manager.create_employee_record("Alice","Marketing",75000)
print("Employee record:", employee)
````

````python
print("Department:",manager.get_department(employee))

updated_record = manager.update_salary(employee, 800000)
print("After salary update:",updated_record)
````

````python
final_record = manager.add_attribute(employee,"location","New York")
print("After adding New Attribute:",final_record)
````
