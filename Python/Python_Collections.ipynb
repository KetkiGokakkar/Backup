**Python Collections - Lists, Tuples, Dictionaries and Sets**

## 1. LIST

```python
# What is a List?
# - A list is an ORDERED, MUTABLE collection of items.
# - Defined using square brackets [ ].
# - Allows duplicate values and can hold mixed data types.
```

```python
# Creating a list
fruits = ["apple", "banana", "cherry", "apple"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "two", 3.0, True]

print("Fruits list:", fruits)
print("Numbers list:", numbers)
print("Mixed list:", mixed)
print("Type of fruits:", type(fruits))
```

```python
fruits = ["apple", "banana", "cherry", "apple"]

# index():Returns the index of the first element with the specified value

print("Apple index:",fruits.index("apple"))
print("Cherry index:",fruits.index("cherry"))

#count()	Returns the number of elements with the specified value
print("No of apple:",fruits.count("apple"))  # string must match completely    
print("No of app:",fruits.count("app")) 
```

```python
# Indexing and Negative Indexing
fruits = ["apple", "banana", "cherry", "date"]

print("First item:", fruits[0])
print("Last item:", fruits[-1])
print("Second last item:", fruits[-2])
```

```python
# Slicing a list -> list[start:stop:step]
numbers = [10, 20, 30, 40, 50, 60]

print("First 3 items:", numbers[:3])
print("Items from index 2 to 4:", numbers[2:5:2])
print("Last 2 items:", numbers[-2:])
print("Every 2nd item:", numbers[::2])
print("Reversed list:", numbers[::-1])
print("Items from index 6 to 3:", numbers[6:2:-2])
```

```python
# Adding elements: append() vs insert()
fruits = ["apple", "banana"]

# append() adds a single item at the end
fruits.append("cherry")
print("After append:", fruits)

# insert(index, item) adds an item at a specific position
fruits.insert(1, "mango")
print("After insert at index 1:", fruits)
```

```python
# extend() vs append()
# append() adds the whole object as ONE element
# extend() adds each element of the iterable individually

list1 = [1, 2, 3]
list2 = [4, 5, 6]

list1_copy = list1.copy()
list1_copy.append(list2)
print("Using append:", list1_copy)   # list2 added as a single nested list

list1.extend(list2)
print("Using extend:", list1)        # elements of list2 added individually
```

```python
# Removing elements: remove(), pop(), del, clear()
fruits = ["apple", "banana", "cherry", "banana"]

fruits.remove("banana")          # removes the FIRST matching value
print("After remove('banana'):", fruits)

fruits = ["apple", "banana", "cherry", "banana"]
popped_item = fruits.pop()       # removes & returns the LAST item
print("Popped item:", popped_item)
print("After pop():", fruits)

fruits = ["apple", "banana", "cherry", "banana"]
popped_index = fruits.pop(0)     # removes & returns item at given index
print("Popped item at index 0:", popped_index)
print("After pop(0):", fruits)

fruits = ["apple", "banana", "cherry", "banana"]
del fruits[0]                    # deletes item at given index
print("After del fruits[0]:", fruits)

fruits = ["apple", "banana", "cherry", "banana"]
fruits.clear()                   # empties the list
print("After clear():", fruits)
```

```python
# Sorting a list: sort() : sorts the list ascending by default.
#SYNTAX :->> list.sort(reverse=True|False, key=myFunc)
numbers = [5, 2, 9, 1, 7]
numbers.sort()
print(numbers)
# sort() sorts the list IN PLACE, returns None
numbers.sort()
print("After sort():", numbers)

numbers.sort(reverse=True)
print("Sorted in descending order:", numbers)


cars = ['Ford', 'BMW', 'bmw','volvo']
cars.sort(reverse=False)  # sort method is case sensitive
print("After reverse sort:",cars)

# For case incase sensitivity 
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print("Case Insensitive sort:",thislist)

# sort by length 
cars.sort(key=len)
print("After Length sort:",cars)
```

```python

#reverse() method reverses the current sorting order of the elements.
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.reverse()
print(thislist)
```

```python
numbers = [5, 2, 9, 1, 7]
# sorted() returns a NEW sorted list, original stays unchanged
new_list = sorted(numbers)
print("Original list:", numbers)
print("sorted() result:", new_list)

```

```python
#Custom Sorting Method
words = ['banana', 'pie', 'Washington', 'book']
sorted_words = sorted(words, key=len)
print(sorted_words) # Output: ['pie', 'book', 'banana', 'Washington']

sorted_words = sorted(words,key)
sorted_words
```

```python
# List Comprehension
# Syntax: [expression for item in iterable if condition]

squares = [n ** 2 for n in range(1, 6)]
print("Squares:", squares)

even_numbers = [n for n in range(1, 15) if n % 2 == 0]
print("Even numbers:", even_numbers)

fruits = ["apple", "banana", "cherry", "banana","mango"]
upper_fruits = [fruit.upper() for fruit in ["apple", "banana", "cherry"]]
print("Uppercase fruits:", upper_fruits)
```

```python
# Nested Lists (list of lists) - useful to represent matrices/tables
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Matrix:", matrix)
print("Element at row 1, col 2:", matrix[1][2])

for row in matrix:
    print(row)
```

```python
# Common built-in functions used with lists
numbers = [4, 2, 9, 1, 7]

print("Length:", len(numbers))
print("Sum:", sum(numbers))
print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
print("Sorted copy:", sorted(numbers))
```

```python
# Membership, Concatenation and Repetition
fruits = ["apple", "banana", "cherry"]

print("Is 'banana' in fruits?", "banana" in fruits)
print("Is 'mango' in fruits?", "mango" in fruits)

list1 = [1, 2, 3]
list2 = [4, 5, 6]
print("Concatenation (+):", list1 + list2)
print("Repetition (*):", list1 * 3)
```

```python
# Copy vs Reference (a common list pitfall)
original = [1, 2, 3]

reference = original          # both names point to the SAME list
copy_list = original.copy()   # a NEW, independent list

reference.append(4)
copy_list.append(99)

print("Original:", original)     # changed because 'reference' points to same object
print("Reference:", reference)
print("Copy:", copy_list)        # unaffected
```

## 2. TUPLE

```python
# What is a Tuple?
# - A tuple is an ORDERED, IMMUTABLE collection of items.
# - Defined using parentheses ( ).
# - Allows duplicate values and mixed data types.
# - Because it is immutable, it is faster than a list and can be used as a dict key.
```

```python
# Creating a tuple
colors = ("red", "green", "blue")
numbers = (1, 2, 3, 4)
mixed = (1, "two", 3.0)

# A single-element tuple needs a trailing comma
single = (5,)
not_a_tuple = (5)

print("Colors tuple:", colors)
print("Type of colors:", type(colors))
print("Type of single:", type(single))
print("Type of not_a_tuple:", type(not_a_tuple))
```

```python
# Tuples are IMMUTABLE - trying to modify one raises an error
colors = ("red", "green", "blue")

try:
    colors[0] = "black"
except TypeError as e:
    print("Error:", e)
```

```python
# Indexing and Slicing (works just like lists)
colors = ("red", "green", "blue", "yellow", "purple")

print("First item:", colors[0])
print("Last item:", colors[-1])
print("Slice [1:4]:", colors[1:4])
print("Reversed:", colors[::-1])
```

```python
# Tuple methods: count() and index()
numbers = (1, 2, 3, 2, 4, 2, 5)

print("Count of 2:", numbers.count(2))
print("Index of first 3:", numbers.index(3))
```

```python
# Tuple Packing and Unpacking
# Packing: values are packed into a tuple automatically
person = "Alice", 25, "Engineer"
print("Packed tuple:", person)

# Unpacking: tuple values assigned to separate variables
name, age, profession = person
print("Name:", name)
print("Age:", age)
print("Profession:", profession)

# Unpacking with * to collect remaining items
first, *rest = (1, 2, 3, 4, 5)
print("First:", first)
print("Rest:", rest)
```

```python
# Nested Tuples
nested = (1, (2, 3), (4, 5, 6))

print("Nested tuple:", nested)
print("Accessing nested element:", nested[1][0])
```

```python
# Tuple vs List
# Feature             List                      Tuple
# Mutability         Mutable                   Immutable
# Syntax             [ ]                       ( )
# Performance        Slower                    Faster
# Use as dict key    Not allowed               Allowed (if elements are hashable)

# Converting between list and tuple
sample_list = [1, 2, 3]
sample_tuple = tuple(sample_list)
back_to_list = list(sample_tuple)

print("List to tuple:", sample_tuple, type(sample_tuple))
print("Tuple to list:", back_to_list, type(back_to_list))
```

```python
# Why use a tuple? -- Example: using a tuple as a dictionary key
# (a list CANNOT be used as a dict key because lists are mutable/unhashable)

locations = {
    (12.97, 77.59): "Bengaluru",
    (28.61, 77.20): "Delhi"
}

print("Location lookup:", locations[(12.97, 77.59)])
```

## 3. DICTIONARY

```python
# What is a Dictionary?
# - A dictionary stores data as KEY-VALUE pairs.
# - Defined using curly braces { } with key: value syntax.
# - Keys must be unique and immutable (string, number, tuple).
# - Mutable, and maintains insertion order (Python 3.7+).
```

```python
# Creating a dictionary and accessing values
student = {
    "name": "Alice",
    "age": 22,
    "course": "Data Science"
}

print("Student dict:", student)
print("Name:", student["name"])

# Using get() is safer - returns None (or a default) instead of an error
print("Grade (with get):", student.get("grade"))
print("Grade (with default):", student.get("grade", "Not Assigned"))
```

```python
# Adding and Updating key-value pairs
student = {"name": "Alice", "age": 22}

student["age"] = 23              # update existing key
student["course"] = "Data Science"  # add a new key

print("Updated student:", student)
```

```python
# Removing items: pop(), popitem(), del, clear()
student = {"name": "Alice", "age": 23, "course": "Data Science", "city": "Pune"}

removed_value = student.pop("city")     # removes given key, returns its value
print("Removed value:", removed_value)
print("After pop('city'):", student)

last_item = student.popitem("city")           # removes the LAST inserted item
print("Removed last item:", last_item)
print("After popitem():", student)

del student["age"]                      # deletes a specific key
print("After del student['age']:", student)

student.clear()                         # empties the dictionary
print("After clear():", student)
```

```python
# Dictionary methods: keys(), values(), items()
student = {"name": "Alice", "age": 23, "course": "Data Science"}

print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
```

```python
# Iterating over a dictionary
student = {"name": "Alice", "age": 23, "course": "Data Science"}

print("Iterating over keys:")
for key in student:
    print(key)

print("\nIterating over key-value pairs:")
for key, value in student.items():
    print(key, ":", value)
```

```python
# Dictionary Comprehension
squares = {n: n ** 2 for n in range(1, 6)}
print("Squares dict:", squares)

# Filtering with a condition
even_squares = {n: n ** 2 for n in range(1, 11) if n % 2 == 0}
print("Even squares dict:", even_squares)
```

```python
# Nested Dictionary
students = {
    "student1": {"name": "Alice", "age": 22},
    "student2": {"name": "Bob", "age": 24}
}

print("Nested dictionary:", students)
print("Student1 name:", students["student1"]["name"])
```

```python
# update() - merging dictionaries
defaults = {"theme": "light", "font_size": 12}
user_prefs = {"font_size": 16, "language": "en"}

defaults.update(user_prefs)
print("Merged dictionary:", defaults)
```

```python
# setdefault() - get a value, and insert a default if the key is missing
student = {"name": "Alice"}

age = student.setdefault("age", 18)
print("Age after setdefault:", age)
print("Student dict now:", student)
```

```python
# Membership test on a dictionary -- checks the KEYS by default
student = {"name": "Alice", "age": 23}

print("Is 'name' a key?", "name" in student)
print("Is 'Alice' a key?", "Alice" in student)
print("Is 'Alice' a value?", "Alice" in student.values())
```

## 4. SET

```python
# What is a Set?
# - A set is an UNORDERED collection of UNIQUE items.
# - Defined using curly braces { } (without key:value pairs).
# - Automatically removes duplicate values.
# - Mutable, but elements themselves must be immutable/hashable.
```

```python
# Creating a set
fruits = {"apple", "banana", "cherry", "apple"}  # duplicate "apple" is dropped
print("Fruits set:", fruits)
print("Type of fruits:", type(fruits))

# Empty set trap: {} creates an empty DICT, not a set
empty_dict = {}
empty_set = set()
print("Type of {}:", type(empty_dict))
print("Type of set():", type(empty_set))
```

```python
# Adding elements: add() and update()
fruits = {"apple", "banana"}

fruits.add("cherry")               # add a single element
print("After add:", fruits)

fruits.update(["mango", "grape"])  # add multiple elements from an iterable
print("After update:", fruits)
```

```python
# Removing elements: remove(), discard(), pop(), clear()
fruits = {"apple", "banana", "cherry", "mango"}

fruits.remove("banana")     # removes item; raises KeyError if not found
print("After remove('banana'):", fruits)

fruits.discard("kiwi")      # removes item if present; NO error if missing
print("After discard('kiwi'):", fruits)

popped = fruits.pop()       # removes & returns an ARBITRARY (any) element (sets are unordered)
#if set is empty it will raise error 
print("Popped element:", popped)
print("After pop():", fruits)

fruits.clear()               # empties the set
print("After clear():", fruits)
```

```python
# Set Operations: union() and intersection()
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("Union (A | B):", set_a | set_b)
print("Union method:", set_a.union(set_b))

print("Intersection (A & B):", set_a & set_b)
print("Intersection method:", set_a.intersection(set_b))
```

```python
# Set Operations: difference() and symmetric_difference()
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("Difference A - B:", set_a - set_b)
print("Difference method:", set_a.difference(set_b))

print("Symmetric Difference (A ^ B):", set_a ^ set_b)
print("Symmetric difference method:", set_a.symmetric_difference(set_b))
```

```python
# Set Comprehension
squares = {n ** 2 for n in range(1, 6)}
print("Squares set:", squares)

# Removing duplicates from a list using a comprehension
numbers = [1, 2, 2, 3, 3, 3, 4]
unique_numbers = {n for n in numbers}
print("Unique numbers:", unique_numbers)
```

```python
# frozenset - an immutable version of a set
fs = frozenset([1, 2, 3, 4])
print("Frozenset:", fs)
print("Type:", type(fs))

try:
    fs.add(5)
except AttributeError as e:
    print("Error:", e)
```

```python
# Membership test with a set (very fast - O(1) average time)
fruits = {"apple", "banana", "cherry"}

print("Is 'banana' in fruits?", "banana" in fruits)
print("Is 'mango' in fruits?", "mango" in fruits)
```

```python
# Practical use case: removing duplicates from a list using set()
numbers = [1, 2, 2, 3, 4, 4, 4, 5]

unique_numbers = list(set(numbers))
print("Original list:", numbers)
print("List with duplicates removed:", sorted(unique_numbers))
```
