```python
values = ["Hello", "World", "from", "LTIMindtree"]
result = "*".join(values)

print(result)
# Output: Hello-World-from-LTIMindtree
```

```python
# str.join(iterable)
join() is a string method that takes an iterable (like list, tuple, etc.) and concatenates its elements with the string as a separator.
Unlike print(sep=...), join() actually builds a new string instead of just formatting for printing.
Use join() when you want to construct a single string instead of just printing. --> (e.g., storing, returning, logging)
```

```python
# Option - 1: To use Sep only for first 2 values: join for partial separator
# Here join() is flexible – you can apply "-" only to the first two and " " to the rest.

values = ["Hello", "World", "from", "LTIMindtree"]
result = "-".join(values[:2]) + " " + " ".join(values[2:])

print(result)
```

```python
# Option - 2: To use Join and Sep -- Mixing both:
values = ["Hello", "World", "from", "LTIMindtree"]
print("-".join(values[:2]),*values[2:],sep = " ")
```

```python
# Explanation:

"-".join(values[:2]) → "Hello-World"
*values[2:] → expands to "from", "LTIMindtree"
sep=" " makes print join these parts with spaces.

# Use sep → when directly printing multiple values with the same separator.
# Use join → when you need to create a new string (or bapply different separators in different places).
```

```python
# str.split() - Usage:
    split() is a string method that splits a string into a list based on a separator.
    Default separator = any whitespace (" ", \t, \n, etc.).

# Syntax: string.split(separator, maxsplit)
    separator (optional): The delimiter to split on. Default is whitespace.
    maxsplit (optional): Maximum number of splits. Default = no limit.
```

```python
# Example for split usage: (split by whitespace)
text = "Hello world from LTIM"
words = text.split()
print(words)

# Using Specific identifier: custom separator
text1 = "Hello, world, from, LTIM"
words1 = text1.split(",")
print(words1)

text2 = "Hello-World-from-LTIM"
words2 = text2.split("-")
print(words2)

# Limit the no.of Splits: Limiting splits with maxsplit
text = "Hello-World-from-LTIM"
parts = text.split("-", 1)
print(parts)

text3 = "Hello world from-LTIM"
words3 = text2.split(" ",2)
print(words3)
```

```python
values = ["hello", "world", "from", "LTIM"]

# Print the parts with the required separators
print("-".join(values[:2]), "-".join(values[2:]),sep=" ")
```

```python
# Combine split() + join()
# You can split a string into parts, manipulate, and then join() back.

text = "Hello-World-from-LTIM"
parts = text.split("-")
print(parts)

result = " ".join(parts[:2]) + " - " + parts[3]
print(result)
```

```python
# split() vs join():

# Feature           split()                                 join()

# Purpose           Breaks a string into a list             Combines a list into a string
# Input             String                                  Iterable (list, tuple, etc.)
# Output            List of strings                         Single string
# Example           "a-b-c".split("-") → ['a','b','c']      "-".join(['a','b','c']) → 'a-b-c'

# When to use:
**************
# split() = break string into pieces
# join()  = glue list elements into one string
```

```python
# Example 4: range
# The range function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and stops before a specified number.
```

```python
# Using range function
# Range with stop value
print("Range with stop value:")
for i in range(5):
    print(i, end=" ")  # Output: 0 1 2 3 4
print()
# Range with start and stop values
print("Range with start and stop values:")
for i in range(2, 6):
    print(i, end=" ")  # Output: 2 3 4 5
print()
# Range with start, stop, and step values
print("Range with start, stop, and step values:")
for i in range(1, 10, 2):
    print(i, end=" ")  # Output: 1 3 5 7 9
print()
# Using range to create a list
range_list = list(range(5))
print("List created using range:", range_list)
```

```python
# Example programs in Python that demonstrate type casting, which involves converting a value from one data type to another.¶
```

```python
# Integer to string type casting
num = 100
num_str = str(num)

print("Integer:", num)
print("String:", num_str)
print("Type of num:", type(num))
print("Type of num_str:", type(num_str))

# String to integer type casting
num_str = "123"
num = int(num_str)

print("String:", num_str)
print("Integer:", num)
print("Type of num_str:", type(num_str))
print("Type of num:", type(num))


# String to float type casting
float_str = "456.78"
float_num = float(float_str)

print("String:", float_str)
print("Float:", float_num)
print("Type of float_str:", type(float_str))
print("Type of float_num:", type(float_num))


# Float to integer type casting
float_num = 123.45
int_num = int(float_num)

print("Float:", float_num)
print("Integer:", int_num)
print("Type of float_num:", type(float_num))
print("Type of int_num:", type(int_num))

# List to tuple type casting
list_example = [1, 2, 3, 4, 5]
tuple_example = tuple(list_example)

print("List:", list_example)
print("Tuple:", tuple_example)
print("Type of list_example:", type(list_example))
print("Type of tuple_example:", type(tuple_example))

# Tuple to list type casting
tuple_example = (6, 7, 8, 9, 10)
list_example = list(tuple_example)

print("Tuple:", tuple_example)
print("List:", list_example)
print("Type of tuple_example:", type(tuple_example))
print("Type of list_example:", type(list_example))


# List to set type casting
list_example = [1, 2, 3, 4, 5, 1, 2]
set_example = set(list_example)

print("List:", list_example)
print("Set:", set_example)
print("Type of list_example:", type(list_example))
print("Type of set_example:", type(set_example))

# Set to list type casting
set_example = {6, 7, 8, 9, 10}
list_example = list(set_example)

print("Set:", set_example)
print("List:", list_example)
print("Type of set_example:", type(set_example))
print("Type of list_example:", type(list_example))


# Dictionary keys and values to list type casting
dict_example = {"name": "DK", "age": 25, "city": "New York"}

keys_list = list(dict_example.keys())
values_list = list(dict_example.values())

print("Dictionary:", dict_example)
print("Keys list:", keys_list)
print("Values list:", values_list)
print("Type of keys_list:", type(keys_list))
print("Type of values_list:", type(values_list))
```

```python
# 4. Assignment Operators

# Assignment operators example
a = 10
print("Initial value of a:", a)  # Output: 10
# Add and assign -- Adds 5 to the current value of a (10 + 5)
a += 5
print("After a += 5:", a)  # Output: 15   # same as a = a + 5
# Subtract and assign -- Subtracts 3 from a (15 - 3)
a -= 3
print("After a -= 3:", a)  # Output: 12   # same as a = a - 3
# Multiply and assign -- Multiplies a by 2 (12 * 2)
a *= 2
print("After a *= 2:", a)  # Output: 24  # same as a = a * 2
# Divide and assign
a /= 4
print("After a /= 4:", a)  # Output: 6.0  # same as a = a / 4
# Modulus and assign -- Finds the remainder of a / 3 (6.0 % 3)
a %= 3
print("After a %= 3:", a)  # Output: 0.0  # same as a = a % 3

# Exponentiation and assign -- Computes 2³ = 8
a = 2
a **= 3
print("After a **= 3:", a)  # Output: 8   # same as a = a ** 3
# Floor division and assign -- Floor division gives the quotient without decimals
a //= 3
print("After a //= 3:", a)  # Output: 2   # same as a = a // 3
```

```python
# Nested if statement example
x = 15
if x > 10:
    print("x is greater than 10")
    if x > 20:
        print("x is also greater than 20")
    else:
        print("x is not greater than 20")
else:
    print("x is 10 or less")
# Output:
# x is greater than 10
# x is not greater than 20
```

```python
# Looping statements in Python are used to iterate over a sequence of items or
execute a block of code repeatedly until a certain condition is met.
Here are examples of different looping statements in Python:
```

```python
# Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# Iterating over a string
message = "Hello"
for char in message:
    print(char)
```

```python
# Using range() in a for loop
for i in range(15):
    print(i)
```

```python
# Example of a while loop
count = 0
while count <= 5:
    print(f"Count is: {count}")
    count += 1
```

```python
# Using break to exit the loop early
numbers = [1, 2, 3, 4, 5]
for num in numbers:
    if num == 4:
        break
    print(num)
```

```python
# Using continue to skip an iteration
numbers = [1, 2, 3, 4, 5]
for num in numbers:
    if num == 3:
        continue
    print(num)
```

```python
# Using pass as a placeholder
for i in range(3):
    pass
# The pass statement does nothing, it's often used as a placeholder where code will eventually go.
```

```python
# Nested for loop example
for i in range(3):
    for j in range(2):
        print(f"({i}, {j})")
```
