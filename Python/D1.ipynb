```python
values = ["hello","Python","Programming"]
result ="*".join(values)
print(result)# here * is separater
```

```python
values = ["hello","to","Python","Programming"]
result ="*".join(values[:2]) + " "+ "-".join(values[2:])
print(result)# here * is separater
```

```python
res_split = result.split("*") # here * is the splitter case 
res_split                     # default separator is ' '  i.e. space 
```
