# 1. MRO (Method Resolution Order):

# 1. MRO (Method Resolution Order):
What is MRO?

    MRO determines the order in which classes are searched for a method when multiple inheritance is used.
    Python follows the C3 Linearization (Depth-First, Left-to-Right) rule to resolve methods.

# Method Resolution Order (MRO):

In Python, Method Resolution Order(MRO) is the order by which Python looks for a method or attribute. 
First, the method or attribute is searched within a class, and then it follows the order we specified while inheriting.

In multiple inheritance, the following search order is followed.

First, it searches in the current parent class if not available, then searches in the parents class specified while inheriting (that is left to right.)
We can get the MRO of a class. For this purpose, we can use either the mro attribute or the mro() method.

```python
#  Example for MRO Usgae:
class A:
    def show(self):
        print("Class A")

class B(A):
    def show(self):
        print("Class B")

class C(A):
    def show(self):
        print("Class C")

class D(B, C):  # Multiple Inheritance
    pass

obj = D()
obj.show()  # Which method will be called?

print(D.__mro__)  # MRO Order
```

# Explanation:
The method is searched in D → B → C → A → object.
    Since B appears first, B's show() method is called.

# Checking MRO:
Use ClassName.__mro__ or help(ClassName) to see the method resolution order.

# Single Inheritance
- In single inheritance, a child class inherits from a single-parent class. Here is one child class and one parent class.

```python
# Example - 1: Single Inheritance:
# Parent class
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Employee ID: {self.emp_id}")

# Child class (inherits from Employee)
class Manager(Employee):
    def __init__(self, name, emp_id, department):
        # Call the constructor of the parent class
        super().__init__(name, emp_id)
        self.department = department

    def display_manager_details(self):
        self.display_details()
        print(f"Department: {self.department}")

# Using the classes
mgr = Manager("Dinesh", "M102", "Sales")
mgr.display_manager_details()
```

```python
# Employee.mro()
Manager.mro()
```

# Multiple Inheritance:
In multiple inheritance, one child class can inherit from multiple parent classes. So here is one child class and multiple parent classes.

```python
# Example - 1: Multiple Inheritance Example:
# Parent class 1
class Family:
    def person_info(self, name, age):
        print('Inside Family class')
        print('Name:', name, 'Age:', age)

# Parent class 2
class Company:
    def company_info(self, company_name, location):
        print('Inside Company class')
        print('Name:', company_name, 'location:', location)

# Child class
class Employee(Family, Company):
    def Employee_info(self, salary, skill):
        print('Inside Employee class')
        print('Salary:', salary, 'Skill:', skill)

# Create object of Employee
emp = Employee()

# access data
emp.Employee_info(12000, 'Machine Learning')
emp.person_info('DK', 28)
emp.company_info('Google', 'Atlanta')
```

```python
Employee.mro() # Method Resolution Order
```

```python
# Full Example: Multiple + Multilevel Inheritance with super() and MRO
class Base:
    def __init__(self):
        print("Base constructor called")
        super().__init__()

class A(Base):
    def __init__(self):
        print("A constructor called")
        super().__init__()

class B(A):
    def __init__(self):
        print("B constructor called")
        super().__init__()

class C(A):
    def __init__(self):
        print("C constructor called")
        super().__init__()

class D(B, C):
    def __init__(self):
        print("D constructor called")
        super().__init__()

# Instantiate D
obj = D()

# Print MRO
print("\nMethod Resolution Order:")
for cls in D.__mro__:
    print(cls)
```

```python
# What's Happening:
    D() is called → triggers D.__init__()

        super() moves to B.__init__()
        Then to C.__init__()
        Then to A.__init__()
        Then to Base.__init__()
        Finally to Python’s base object class

Each class cooperates by using super().__init__()
Python walks the class hierarchy using the C3 Linearization algorithm (used in MRO)
```

# Multilevel inheritance
- In multilevel inheritance, a class inherits from a child class or derived class. 
Suppose three classes A, B, C. A is the superclass, B is the child class of A, C is the child class of B. 
In other words, we can say a chain of classes is called multilevel inheritance.

```python
# Example - 1: Multilevel Inheritance Example:
# Base class
class Grandparents:
    def bunglow(self):
        return 'Goa Bunglow'

# Child class
class Parents(Grandparents):
    def car(self):
        return 'BMW Car'

# Child class
class You(Parents):
    def phone(self):
        return 'Iphone 19'

current_gen = You()
print(current_gen.bunglow(), current_gen.car(), current_gen.phone())
g = You()
print(g.phone())
print(g.bunglow())
print(g.car())
```

```python
dad = Parents()
dad.bunglow(), dad.car()
```

```python
dad.phone()
```

```python
# Example - 2: Banking System with Multilevel & Multiple Inheritance (Combined)
# Parent Class
class Person:
    def __init__(self, name):
        self.name = name

# Intermediate Class
class BankCustomer(Person):                                   # BankCustomer inherits from Person (Multilevel)
    def __init__(self, name, account_number):
        super().__init__(name)
        self.account_number = account_number

# Another Parent Class
class KYC:
    def verify_identity(self):
        print("KYC verification completed.")

# Final Class using Multilevel + Multiple inheritance
class LoanCustomer(BankCustomer, KYC):                        # LoanCustomer inherits from BankCustomer (Multilevel) 
    def __init__(self, name, account_number, loan_amount):    # LoanCustomer inherits from both BankCustomer and KYC(Multiple)
        super().__init__(name, account_number)
        self.loan_amount = loan_amount

    def show_details(self):
        print(f"Customer Name: {self.name}")
        print(f"Account Number: {self.account_number}")
        print(f"Loan Approved: ₹{self.loan_amount}")
        self.verify_identity()

# Using the class
lc = LoanCustomer("DK", "ACC987654", 500000)
lc.show_details()
```

```python
LoanCustomer.mro()
```

# Hierarchical Inheritance
- One parent class and multiple child classes.

```python
# In Hierarchical Inheritance, multiple child classes inherit from a single parent class.
# Business Example: Employee is the parent class. Manager and Clerk are child classes.

# Parent class
class Employee:                                              # Employee is the base class.
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def show_details(self):
        print(f"Employee ID: {self.emp_id}, Name: {self.name}")

# Child class 1
class Manager(Employee):                                    # Manager and Clerk both inherit from Employee.
    def __init__(self, name, emp_id, department):           # Each child class has specific responsibilities.
        super().__init__(name, emp_id)
        self.department = department

    def assign_task(self):
        print(f"Manager {self.name} is assigning tasks in {self.department} department.")

# Child class 2
class Clerk(Employee):
    def __init__(self, name, emp_id, section):
        super().__init__(name, emp_id)
        self.section = section

    def file_documents(self):
        print(f"Clerk {self.name} is filing documents in {self.section} section.")

# Using the classes
manager1 = Manager("DK", "M1001", "Finance")
clerk1 = Clerk("AK", "C2001", "Records")

# Showing details and behaviors
manager1.show_details()
manager1.assign_task()

print("-----")

clerk1.show_details()
clerk1.file_documents()
```

# 3. Constructors and Destructors in Python (OOPs):

# Constructors and Destructors in Python (OOPs):

In Object-Oriented Programming (OOPs), a constructor is a special method used to initialize objects, 
while a destructor is used to perform cleanup operations when an object is no longer in use.

# 1. Constructor in Python
A constructor is a special method that is automatically called when an object of a class is created. It is defined using the __init__() method.

# Types of Constructors:
    1. Default Constructor – No parameters except self
    2. Parameterized Constructor – Takes arguments for initialization

```python
# Default Constructor Example: No parameters except self:
class Student:
    def __init__(self):   # Default constructor (no parameters except self)
        self.name = "NA"
        self.age = 18
    
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")

# Creating object
s1 = Student()
s1.display()

s2 = Student()
s2.display()

s3 = Student()
s3.display()
```

```python
# Parameterized Constructor (Takes arguments for initialization):
class Student:
    def __init__(self, name, age):   # Parameterized constructor
        self.name = name
        self.age = age
    
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")

# Creating objects with different values
s1 = Student("Ajay", 20)
# s2 = Student()

s1.display()
# s2.display()
```

```python
# Combined Constructor Example: Defualt and Parametrized:
class Student:
    # Constructor with default values
    def __init__(self, name="Unknown", age=18):
        self.name = name
        self.age = age
    
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")

# Object without arguments → uses default values
s1 = Student()
s1.display()

# Object with custom values → parameterized
s2 = Student("Ajay", 20)
s2.display()

# Object with only one argument → mixes default and parameterized
s3 = Student("Sneha")
s3.display()
```

```python
# 2. Destructor in Python:
# A destructor is a special method that is called when an object is deleted or goes out of scope. 
# It is defined using the __del__() method.

# Example of a Destructor:

class Car:
    def __init__(self, brand):
        self.brand = brand
        print(f"Car {self.brand} is created.")

    def __del__(self):
        print(f"Car {self.brand} is being destroyed.")

# Creating an object
c1 = Car("Tesla")

# Deleting the object explicitly
del c1  # Make sure there's no unnecessary space before this line
```

# Explanation:
 - __init__() initializes the object with a brand name.
 - __del__() is called when c1 is deleted using del c1.
 - This ensures proper resource management, such as closing files or network connections.

# Summary:
    - Constructor (__init__()) initializes an object.
    - Destructor (__del__()) is called when an object is destroyed.
    - Explicit del object triggers the destructor immediately.
    - Python’s garbage collector automatically calls __del__() when an object is no longer referenced.

# Understanding Method Overriding in Python:

### 1. **Method Overriding**

- **Purpose:** Method overriding allows a subclass to provide a new implementation for a method that is already defined in its parent (super) class.
- **How it works:** In method overriding, the method in the child class should have the same name, signature (parameters), and return type as the method in the parent class. The child class can then define its own version of the method, which will be called instead of the method in the parent class when invoked on an instance of the subclass.

**Method overriding is an object-oriented programming feature that allows a subclass to provide a specific implementation of a method that is already defined in its parent class. The overridden method in the subclass must have the same name, parameters, and return type as the method in the parent class.**

```python
# Example: Overriding Interest Calculation in Different Bank Accounts:
# Parent class:
class BankAccount:
    def interest_rate(self):
        return "The interest rate for a general bank account is 3%."

# Child class: Savings Account:
class SavingsAccount(BankAccount):
    def interest_rate(self):
        return "The interest rate for a savings account is 5%."

# Child class: Fixed Deposit Account:
class FixedDepositAccount(BankAccount):
    def interest_rate(self):
        return "The interest rate for a fixed deposit account is 7%."

# Creating instances of each account type:
general_account = BankAccount()
savings_account = SavingsAccount()
fixed_deposit_account = FixedDepositAccount()

# Calling the overridden method:
print(general_account.interest_rate())          # Output: The interest rate for a general bank account is 3%.
print(savings_account.interest_rate())          # Output: The interest rate for a savings account is 5%.
print(fixed_deposit_account.interest_rate())    # Output: The interest rate for a fixed deposit account is 7%.

# Explanation:
# -- BankAccount (Parent class): Provides a general interest rate of 3%.
# -- SavingsAccount (Child class): Overrides interest_rate() and sets it to 5%.
# -- FixedDepositAccount (Child class): Overrides interest_rate() and sets it to 7%.
# When we call interest_rate() on different account objects, the overridden method in the subclass executes instead of the parent class’s method.
```

```python
# Example - 2:

# Parent Class
class BankAccount:                                      # BankAccount is a parent/base class
    def __init__(self, account_holder, balance):        # __init__() initializes the account holder name and balance.
        self.account_holder = account_holder
        self.balance = balance

    def calculate_interest(self):               # calculate_interest() is a method that just prints a message for basic accounts (no interest applied).
        print(f"[BankAccount] No interest applied for basic accounts.")

class SavingsAccount(BankAccount):         # Child Class - Savings Account / SavingsAccount inherits from BankAccount.
    def calculate_interest(self):          # It overrides the calculate_interest() method with its own logic.
        #super().calculate_interest()       # Calls the parent class method/logic (optional)
        interest = self.balance * 0.04     # Calculates 4% interest.
        print(f"[SavingsAccount] Interest: ${interest:.2f}")  # Prints the interest with formatted output (.2f ensures 2 decimal places).
        print(f"[SavingsAccount] Interest for {self.account_holder}: ${interest:.2f}")
        
account1 = SavingsAccount("Priya", 10000)
account1.calculate_interest()

account2 = BankAccount("DK",5000)
account2.calculate_interest()
```

# Understanding Method Overloading in Python:

# Understanding Method Overloading in Python:
In traditional OOP languages (like Java or C++), method overloading means defining multiple methods with the same name but different parameters  (type, number, or order).

# BUT in Python…

Python does not support traditional method overloading directly.
Why? Because Python is dynamically typed — it allows flexible function arguments using default values, *args, and **kwargs.

# Instead, we simulate overloading.

# Simulating Method Overloading in Python:
# We do this by:
    Using default arguments
    Using *args for variable number of arguments / Variable arguments (*args, **kwargs)
    Custom logic inside method to behave differently / Type checking (e.g., isinstance)

```python
# Example-1: E-Commerce Discount System
```

```python
# Imagine an E-commerce application where customers get discounts in different ways:
    A flat discount (e.g., ₹500 off).
    A percentage discount (e.g., 10% off).
    A coupon-based discount (code like "NEW10").
Here, instead of creating multiple methods (apply_flat_discount, apply_percentage_discount, apply_coupon_discount),
we overload a single method apply_discount() that behaves differently depending on arguments.
```

```python
class Discount:        # Declares a new class named Discount. A class groups related data and behavior (methods).
    def apply_discount(self, *args):
        if len(args) == 1 and isinstance(args[0], int):
            # Case 1: Flat discount
            amount = args[0]
            return f"Applied flat discount of ₹{amount}"
        
        elif len(args) == 1 and isinstance(args[0], float):
            # Case 2: Percentage discount
            percent = args[0] * 100
            return f"Applied {percent}% discount"
        
        elif len(args) == 2 and isinstance(args[0], str) and isinstance(args[1], int):
            # Case 3: Coupon discount
            coupon, amount = args
            return f"Applied coupon '{coupon}' for ₹{amount} off"      
        else:
            return "Invalid discount type"
```

```python
# --- Usage ---
d = Discount()

print(d.apply_discount(500))           # Flat discount
print(d.apply_discount(0.10))          # Percentage discount
print(d.apply_discount("NEW10", 1000))  # Coupon discount
```

# Business Understanding

# Problem without overloading: 
You’d need separate methods (apply_flat_discount(), apply_percentage_discount(), apply_coupon_discount()), 
making your API cluttered.

# With overloading:

A single unified method name (apply_discount) is used.
Code is cleaner and easier to maintain.
Developers just remember one method, and arguments decide behavior.

# In a banking application, customers can deposit money in different ways:

    Cash deposit (direct amount).
    Cheque deposit (cheque number + amount).
    Online transfer (from account + amount).

Instead of writing multiple methods like deposit_cash(), deposit_cheque(), deposit_online(),
we overload a single method deposit().

```python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, *args):
        if len(args) == 1 and isinstance(args[0], int):
            # Case 1: Cash deposit
            amount = args[0]
            self.balance += amount
            return f"Cash deposit of ₹{amount}. New Balance: ₹{self.balance}"

        elif len(args) == 2 and isinstance(args[0], str) and isinstance(args[1], int):
            # Case 2: Cheque deposit
            cheque_no, amount = args
            self.balance += amount
            return f"Cheque {cheque_no} deposit of ₹{amount}. New Balance: ₹{self.balance}"

        elif len(args) == 2 and isinstance(args[0], int) and isinstance(args[1], int):
            # Case 3: Online transfer
            from_acc, amount = args
            self.balance += amount
            return f"Online transfer from A/C {from_acc} of ₹{amount}. New Balance: ₹{self.balance}"

        else:
            return "Invalid deposit method"

# --- Usage ---
acc = BankAccount("Dinesh", 5000)

print(acc.deposit(2000))                 # Cash deposit
print(acc.deposit("CHQ12345", 1500))     # Cheque deposit
print(acc.deposit(987654321, 3000))      # Online transfer
```
