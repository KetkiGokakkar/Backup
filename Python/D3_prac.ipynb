# Question 1 — E-Commerce Order Priority Management

## Business Scenario

An e-commerce company maintains a list containing the **order values of orders waiting for processing**.

During the day:

  * New orders are received.
  * Bulk orders are imported from another system.
  * A VIP order must be placed at a specific position.
  * A cancelled order must be removed.
  * The operations team wants to identify an order's position.
  * Duplicate order amounts must be counted.
  * Orders have to be sorted by value.
  * A backup must be created before clearing the end-of-day processing list.

### Sample Data
orders = [1200, 850, 1500, 600, 850]
new_orders = [2000, 750, 1100]

## Functions to Implement

```python
orders = [1200, 850, 1500, 600, 850] 
new_orders = [2000, 750, 1100]
def add_order(orders, order_amount):
    orders.append(order_amount)

def add_bulk_orders(orders, new_orders):
    orders.extend(new_orders)

def add_vip_order(orders, position, order_amount):
    orders.insert(position,order_amount)

def cancel_order(orders, order_amount):
    orders.remove(order_amount)

def remove_last_order(orders):
    orders.pop()

def find_order_position(orders, order_amount):
    return orders.index(order_amount)

def count_order_amount(orders, order_amount):
    cnt=orders.count(order_amount)
    return cnt

def sort_orders(orders):
    sorted_orders=sorted(orders)
    return sorted_orders
    
def reverse_orders(orders):
    orders.reverse()

def backup_orders(orders):
    backup = orders.copy()
    return backup

def close_day(orders):
    orders.clear()


add_order(orders,950)
print(orders)

add_bulk_orders(orders,new_orders)
print(orders)

add_vip_order(orders,1,950)
print(orders)

cancel_order(orders,600)
print(orders)

remove_last_order(orders)
print(orders)

print(find_order_position(orders,1500))

print(count_order_amount(orders,850))

print(sort_orders(orders))
      
reverse_orders(orders)
print(orders)
reverse_orders(orders)
backup = backup_orders(orders)
close_day(orders)
print(orders)
```

   
## Requirements

  Perform the following business operations:

  1. Add an order worth `950`.
  2. Add bulk orders `[2000, 750, 1100]`.
  3. Insert a VIP order worth `5000` at position `1`.
  4. Cancel the order having value `600`.
  5. Remove the last order from the processing queue.
  6. Find the position of order value `1500`.
  7. Count how many orders have value `850`.
  8. Sort orders in ascending order.
  9. Reverse the sorted order.
  10. Create a backup copy.
  11. Clear the original list after processing.
