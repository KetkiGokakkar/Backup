# SQLite / SQLAlchemy for Database Interaction in Python

    SQLite = the database.
    SQLAlchemy = a Python library that makes database interaction easier and more portable.

# SQLite is lightweight, file-based, and does not require a separate database server. Python includes the sqlite3 DB-API interface in its standard library.

# SQLAlchemy provides a higher-level database abstraction using either SQLAlchemy Core or the ORM. Current SQLAlchemy 2.x applications normally start with an Engine, which manages database connectivity.

# SQLite vs SQLAlchemy
# Feature	              SQLite sqlite3	         SQLAlchemy
What is it?	              Python SQLite interface	 Database toolkit/ORM
SQL knowledge required	  High	                     Core: medium; ORM: less
Write SQL manually	      Usually yes	             Optional
SQLite support	          Yes	                     Yes
PostgreSQL/MySQL etc.	  No	                     Yes
ORM support	              No	                     Yes
Pythonclasses as tables	  No	                     Yes
Connection management	 Manual/basic	Engine/pooling
Good for beginners	     Excellent	    After basic SQL
Good for large apps	     Limited	    Excellent

# SQLAlchemy:

SQLAlchemy 2.x uses select() as the standard query construct, and ORM operations are typically performed through a Session

# Engine → Connection → Model → ORM → Session → select() → add() → commit() → delete()


    The Engine is the main object that knows how to connect to your database.
    A Connection represents an actual database connection obtained through the Engine.
    A Model is a Python class that represents a database table when using SQLAlchemy ORM.
    ORM - Object Relational Mapping -- It Maps Python Classes to Database Tables 
    SQLAlchemy ORM converts the object operation into the appropriate database operation.

# Without ORM:
cursor.execute("""
INSERT INTO students
VALUES (101, 'Arun', 'Python', 12000)
""")

# With ORM:
student = Student(
    student_id=101,
    student_name="Arun",
    course="Python",
    fee=12000
)

    The Session is one of the most important ORM concepts.

    It manages your Python ORM objects and coordinates database operations and transactions. SQLAlchemy describes it as the main interface through     which ORM objects are loaded, modified, persisted, and queried.

```python
# Example - Working with Snowflake - SQLAlchemy:

import pandas as pd
from urllib.parse import quote_plus
from sqlalchemy import create_engine

# Connection details
user = 'DKSF'
password = 'Dk@snowflake27' 
account = 'WXOQYRX-BO26997'
database = 'RETAIL_DW_DB'
schema = 'DW'
warehouse = 'COMPUTE_WH'

quoted_password = quote_plus(password)

# SQLAlchemy engine format:
# snowflake://<user>:<password>@<account>/<database>/<schema>?warehouse=<warehouse>
engine = create_engine((f'snowflake://{user}:{quoted_password}@{account}/{database}/{schema}?warehouse={warehouse}'))

# Query using pandas.read_sql
df = pd.read_sql("SELECT * From RETAIL_DW_DB.DW.DIM_CUSTOMER LIMIT 3", engine)

# Display result
print(df)
```
