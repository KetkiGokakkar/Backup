# 1. Handling Missing Values:
    Identifying Missing Values
    Missing values can occur in any column of the DataFrame. 
    You can identify missing values with isna() or isnull(), 
which return a DataFrame of boolean values indicating where the data is missing.

```python
import pandas as pd
import numpy as np

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve'],
    'Age': [24, np.nan, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', None]
}

df = pd.DataFrame(data)

# Identify missing values
print(df.isna())
```

```python
In a pandas DataFrame, np.nan refers to "Not a Number" (NaN) from the NumPy library. It represents missing or undefined numerical data.
```

# Key Points About np.nan:
     Acts as a placeholder for missing values in numerical columns.
     Propagation in calculations: Most operations involving np.nan will result in np.nan (e.g., 5 + np.nan → np.nan).
     Not equal to itself (np.nan == np.nan is False), so use pd.isna() or pd.isnull() to check for NaN values.
     Can be handled using methods like fillna(), dropna(), etc.

# Handling Missing Values with dropna():
    The dropna() method is used to remove missing values (NaN). You can drop rows or columns that contain missing data.

```python
# Dropping rows with any missing values:
  
import pandas as pd

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve'],
    'Age': [24, None, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', None]
}
df = pd.DataFrame(data)
df_dropped = df.dropna()
print(df_dropped)  
```

```python
# Dropping rows where all columns are NaN: Example-1

import pandas as pd

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve',None],
    'Age':  [24, None, 22, 32, 29,None],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', None,None]
}
df = pd.DataFrame(data)
df_dropped_all = df.dropna(how='all')
print(df_dropped_all)
```

# Filling Missing Values with fillna():  
The fillna() method allows you to replace missing values with a specific value or method, such as using the mean, median, or mode of the column.

```python
# Filling with a constant value:
import pandas as pd

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve'],
    'Age': [24, None, 30, 32, 29],
    'City': ['New York', 'Los Angeles', None, 'Houston', None]
}
df = pd.DataFrame(data)
print(df)

df_filled = df.fillna({'Age': 30, 'City': 'Unknown'})
print(df_filled)
```

```python
# Filling with the mean (for numeric columns)
import pandas as pd

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve'],
    'Age': [24, None, 30, 32, 29],
    'City': ['New York', 'Los Angeles', None, 'Houston', None]
}
df = pd.DataFrame(data)
print(df)

df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

```python
# Filling forward (using the previous value):

import pandas as pd

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve'],
    'Age': [24, None, 30, 32, 29],
    'City': ['New York', 'Los Angeles', None, 'Houston', None]
}
df = pd.DataFrame(data)
print(df)

df_filled_forward = df.ffill()
print(df_filled_forward)
```

```python
# Filling backward (using the next value):

df_filled_backward = df.bfill()
print(df_filled_backward)
```

# 2. Handling Duplicates:

    Identifying Duplicates
    You can check for duplicate rows using the duplicated() method, which returns a boolean Series indicating whether each row is a duplicate.

```python
# Sample DataFrame with duplicates:
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob'],
    'Age': [24, 27, 24, 32, 27],
    'City': ['New York', 'Los Angeles', 'New York', 'Chicago', 'Los Angeles']
}
df = pd.DataFrame(data)

# Check for duplicate rows
print(df.duplicated())
```

# Removing Duplicates:
The drop_duplicates() method removes duplicate rows based on the column(s) you specify.

```python
# Removing all duplicates:
df_no_duplicates = df.drop_duplicates()
print(df_no_duplicates) 
```

# Indexing and Selection in DataFrames :

    o  loc and iloc for label-based and integer-based indexing. 
    o  slicing 
    o  boolean indexing 
    o  Boolean indexing for filtering data.  
  
These operations allow you to access specific rows, columns, or elements, and modify or filter your data efficiently 
  
1. loc and iloc Indexing:
loc is label-based indexing, which means you use labels or index names to access data.
iloc is integer-based indexing, which means you use integer positions to access data (i.e., row/column positions as numbers).

In pandas, .loc[] and .iloc[] are used for data selection and slicing, but they differ in how they reference rows and columns.

1. .loc[] → Label-based indexing:
                                                                     
	•	Uses explicit labels (row or column names).
	•	Can accept boolean arrays, lists of labels, or slicing with labels.
	•	Includes the end index in slicing.

2. .iloc[] → Position-based indexing:

	•	Uses integer positions (like Python lists).
	•	End index in slicing is excluded.
	•	Accepts integer lists, slicing, or boolean arrays.  

```python
# .loc[] → Label-based indexing:

import pandas as pd

# Creating a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'Score': [85, 90, 95, 88]}

df = pd.DataFrame(data, index=['A', 'B', 'C', 'D'])

# Using .loc to select a single row
print(df.loc['B'])  # Select row with label 'B'

# Using .loc to select multiple rows
print(df.loc[['A', 'C']])  # Select rows 'A' and 'C'

# Using .loc for slicing (label-based, includes 'C')
print(df.loc['A':'C'])

# Selecting specific columns with .loc
print(df.loc[:, ['Name', 'Score']])  # Select all rows, but only 'Name' and 'Score' columns 
```
