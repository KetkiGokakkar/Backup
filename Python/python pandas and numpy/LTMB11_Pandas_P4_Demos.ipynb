```python
1. Reshaping and Transforming Data:
  
groupby() for Grouping Data:
-- The groupby() method in Pandas is a powerful tool to group data based on one or more columns. 
-- It can be followed by aggregation functions such as sum(), mean(), or count() to summarize data.  
```

```python
# Example: Grouping by a Column:
    
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['AK', 'BK', 'CK', 'AK', 'BK', 'Ck'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'New York', 'Los Angeles'],
    'Age': [24, 27, 22, 32, 29, 28],
    'Salary': [50000, 60000, 55000, 80000, 75000, 72000]
}
df = pd.DataFrame(data)

print(df)
```

```python
# Group by 'Name' and calculate the average salary for each person
Name_grouped = df.groupby('Name')['Salary'].mean()         # grouped = df.groupby('Name')['Salary'].sum()
print(Name_grouped)

# Group by 'Name' and calculate the average salary for each person
City_grouped = df.groupby('City')['Salary'].sum()         # grouped = df.groupby('City')['Salary'].sum()
print(City_grouped)    
```

```python
# Aggregating Data with groupby():
# You can apply multiple aggregation functions to grouped data using the agg() method.  
  
# Group by 'City' and get mean and sum of 'Age' and 'Salary'
print(df)

grouped_agg = df.groupby('City').agg({'Age': 'mean','Salary': 'sum'})
print(grouped_agg) 
```

# Pivot Tables with pivot_table():
  
The pivot_table() function is used to reshape data by creating a new table where you can specify rows,
columns, and aggregation functions for data transformation.  

```python
# Example - 1:

import pandas as pd

# Sample data
data = {
    'Product': ['Apple', 'Banana', 'Carrot', 'Apple', 'Banana', 'Carrot', 'Apple', 'Carrot'],
    'Category': ['Fruit', 'Fruit', 'Vegetable', 'Fruit', 'Fruit', 'Vegetable', 'Fruit', 'Vegetable'],
    'Quantity Sold': [30, 45, 20, 50, 60, 35, 40, 55],
    'Revenue': [60, 90, 40, 100, 120, 70, 80, 110]
}

# Creating a DataFrame
df = pd.DataFrame(data)

print(df)  
```

```python
# Pivot table for total quantity sold and revenue by category
pivot_table = pd.pivot_table(df, 
                             values=['Quantity Sold', 'Revenue'],                         # Specifies which columns should be aggregated.
                             index='Category',                                            # index defines the grouping column
                             aggfunc={'Quantity Sold': 'sum', 'Revenue': 'sum'})          # aggfunc defines how values should be aggregated.

print(pivot_table)  
```

```python
pivot_table = pd.pivot_table(df, values='Quantity Sold', index='Category', aggfunc='sum')
print(pivot_table) 
```

```python
pivot_table = pd.pivot_table(df, values='Quantity Sold', index='Category', aggfunc='mean')
print(pivot_table)  
```

```python
pivot_table = pd.pivot_table(df, values='Revenue', index='Category', aggfunc='count')
print(pivot_table) 
```

```python
# Multiple Functions:
pivot_table = pd.pivot_table(df, values='Revenue', index='Category', aggfunc=['sum', 'mean'])
print(pivot_table)  
```

```python
# Using Different Aggregations for Different Columns:
pivot_table = pd.pivot_table(df, 
                             values=['Quantity Sold', 'Revenue'], 
                             index='Category', 
                             aggfunc={'Quantity Sold': 'sum', 'Revenue': 'mean'})
print(pivot_table)  
```

```python
# Example with Multiple Aggregation Functions:
pivot_table = pd.pivot_table(df, 
                             values=['Quantity Sold', 'Revenue'], 
                             index='Category', 
                             aggfunc={'Quantity Sold': ['sum', 'mean'], 'Revenue': ['sum', 'max']})
print(pivot_table)  
```

# apply() Function:
The apply() method is used to apply a function along an axis of a DataFrame (either rows or columns). 
You can pass custom functions or use built-in functions.  

```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['AK', 'BK', 'CK', 'Alice', 'BK', 'Ck'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'New York', 'Los Angeles'],
    'Age': [24, 27, 22, 32, 29, 28],
    'Salary': [50000, 60000, 55000, 80000, 75000, 72000]
}
df = pd.DataFrame(data)

print(df)

# Apply a function only for 1 column - Age
numeric_column = df[['Age']].apply(sum)     # Object Type DataFrame  - Works - Column-wise if using with apply() 
print(numeric_column)                       # If we use df['Age'].apply(sum) - error - Goes as series - Performs element-wise Operation-invalid

# Alternate Option:
numeric_column1 = df['Age'].sum()   # Object Type Series Best practice if using wo apply() 
print(numeric_column1)

# Apply a function to each column to get the sum of each column
numeric_column_sums = df.select_dtypes(include='number').apply(sum)       #column_sums = df.apply(sum) 
print(numeric_column_sums)

# Apply concatenation to string columns
string_column_concat = df.select_dtypes(include='object').apply(lambda x: ' '.join(x))
print(string_column_concat) 
```

# map() for Element-wise Transformation:
  
The map() function is used for element-wise transformations, often for transforming or mapping values in a column
(e.g., applying a function to each value or replacing values with a dictionary).  

```python
# Example: Mapping Values Using a Dictionary  
  
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'New York', 'Los Angeles'],
    'Age': [24, 27, 22, 32, 29, 28],
    'Salary': [50000, 60000, 55000, 80000, 75000, 72000]
}
df = pd.DataFrame(data)

print(df)

# Mapping city names to their respective regions:
city_map = {'New York': 'East', 'Los Angeles': 'West', 'Chicago': 'Midwest'}
df['Region'] = df['City'].map(city_map)
print(df) 
```

# 3. Merging and Concatenating DataFrames
Merging DataFrames with merge() Pandas provides a flexible and powerful merge() function to combine two DataFrames based on a common column (similar to SQL joins).  

```python
#Example: Merging DataFrames 
  
# Create two DataFrames with a common column 'Name'

df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [24, 27, 22]
})

df2 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Salary': [50000, 60000, 55000]
})

# Merge DataFrames on 'Name'
merged_df = pd.merge(df1, df2, on='Name')
print(merged_df)  
  
# The merge() function performs an inner join by default, meaning it combines only the rows with matching values in the Name column 
```

```python
  # Left join (all rows from df1(left), matching rows from df2(right))
left_joined = pd.merge(df1, df2, on='Name', how='left')
print(left_joined)  
  
# Right join (all rows from df2(right), matching rows from df1(left))
right_joined = pd.merge(df1, df2, on='Name', how='right')
print(right_joined) 
```

# Concatenating DataFrames with concat():
The concat() function is used to concatenate DataFrames along a particular axis (rows or columns).

```python
# Example: Concatenating DataFrames Vertically

import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'New York', 'Los Angeles'],
    'Age': [24, 27, 22, 32, 29, 28],
    'Salary': [50000, 60000, 55000, 80000, 75000, 72000]
}
df = pd.DataFrame(data)

# Concatenate DataFrames vertically (stacking rows)

df3 = pd.DataFrame({
    'Name': ['Eve', 'David'],
    'Age': [29, 32],
    'Salary': [75000, 80000]
})

concatenated_df = pd.concat([df, df3], axis=0, ignore_index=False)
print(concatenated_df)
```

```python
# Summary:
	If you keep ignore_index=False, the new rows reuse indices 0,1.
	To access them: use .iloc[] (by position), or .loc[] (will return multiple matches).
	If you want continuous unique index, set ignore_index=True  
```

```python
# Example: Concatenating DataFrames Horizontally:
    
# Concatenate DataFrames horizontally (side by side)
df4 = pd.DataFrame({
    'Country': ['USA', 'USA', 'USA', 'USA', 'USA'],
    'Continent': ['North America', 'North America', 'North America', 'North America', 'North America']
})

concatenated_df_horizontal = pd.concat([df, df4], axis=1)
print(concatenated_df_horizontal)
```

# Working with Regular Expressions: re.sub:

re.sub() is used to replace occurrences of a regex pattern with a specified replacement string.

🔹 Syntax: re.sub(pattern, replacement, string, count=0, flags=0)

```python
# Remove Special Characters:  --> [^\w\s] removes everything except letters, digits, and whitespace.
import re

text = "Hello!!! Welcome @ Python #2026"
cleaned = re.sub(r'[^\w\s]', '', text)

print(cleaned)
```

```python
# Remove Numbers: --> \d+ matches one or more digits.
text = "Order ID: 45982, Amount: 3500"
result = re.sub(r'\d+', '', text)

print(result)  
```

```python
# Replace Multiple Spaces with Single Space:  --> \s+ matches multiple whitespace characters.
text = "Python     is    powerful"
result = re.sub(r'\s+', ' ', text)

print(result) 
```

```python
# Remove URLs: --> http\S+ removes URL starting with http.
text = "Visit our site at https://example.com for details"
result = re.sub(r'http\S+', '', text)

print(result)   
```

```python
# Replace Digits with #
text = "My phone number is 9876543210"
result = re.sub(r'\d', '#', text)

print(result)  
```

```python
# Case-Insensitive Replacement:
text = "Python is python and PYTHON"
result = re.sub(r'python', 'Java', text, flags=re.IGNORECASE)

print(result) 
```

```python
# Limit Number of Replacements:

text = "apple apple apple"
result = re.sub(r'apple', 'orange', text, count=2)

print(result)   
```

```python
# Keep Only Alphabets:
text = "Python3.10 is #1!"
result = re.sub(r'[^a-zA-Z]', '', text)

print(result)  
```
