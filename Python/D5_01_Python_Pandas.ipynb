# Accessing Rows by Index: You can use the loc[] method for label-based indexing or iloc[] for position-based indexing.
# Slicing: DataFrames can be sliced by rows or columns.

```python
data = [
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago'],
    ['David', 40, 'Houston']
]

df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])
print(df)

# This returns rows at positions 1 and 2 (i.e., Bob and Charlie). It does not include position 3 (David).
print(df[1:3])  # Slicing rows from index 1 to 2

# Important distinction: df[1:3] vs df.loc[...]:

# df[1:3] - This is called as (positional slice)
# start - included
# stop - excluded

# df.loc[1:3] (label slice, if index is 0..n)
# start included
# stop included (inclusive!)
print(df.loc[1:3])  # This includes 1, 2, and 3 → Bob, Charlie, David.

# And iloc is always stop-exclusive - integer based - start inclusive, stop exclusive (same like positional)
print(df.iloc[1:3]) # Same as df[1:3] → rows 1 and 2 only.
```

# What happens when the index is not 0..n
what changes when you use a custom index (not 0..n)

```python
import pandas as pd

data = [
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago'],
    ['David', 40, 'Houston']
]

df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])

print("=== Original DataFrame (default index) ===")
print(df)

# ---------------------------
# 2) Default index slicing
# ---------------------------
print("\n=== df[1:3] (positional slice: start included, stop excluded) ===")
print(df[1:3])

print("\n=== df.iloc[1:3] (positional slice: start included, stop excluded) ===")
print(df.iloc[1:3])

print("\n=== df.loc[1:3] (label slice: start included, stop INCLUDED) ===")
print(df.loc[1:3])

# ---------------------------
# 3) Custom index example
# ---------------------------
df_custom = df.copy()
df_custom.index = [101, 105, 110, 120] # Ensure the Custom Index is sorted before slicing, else will get unexpected results

print("\n=== DataFrame with Custom Index ===")
print(df_custom)

print("\n=== df_custom[1:3] (still positional, NOT label-based) ===")
print(df_custom[1:3])

print("\n=== df_custom.iloc[1:3] (positional) ===")
print(df_custom.iloc[1:3])

print("\n=== df_custom.loc[105:120] (label-based, stop INCLUDED) ===")
print(df_custom.loc[105:120])
```

# Filtering: You can filter rows based on conditions.

```python
import numpy as np
import pandas as pd

# Define the data
data = np.array([[1, 'Alice', '25'], [2, 'Bob', '30'], [3, 'Charlie', '35']])

# Create the DataFrame
df = pd.DataFrame(data, columns=['ID', 'Name', 'Age'])

# Convert the 'Age' column to integers
df['Age'] = df['Age'].astype(int)

# Print the original DataFrame
print(df)

# Filter rows where Age is greater than 30
print(df[df['Age'] > 30])  # This will work now  -- This filters the DataFrame to include only the rows where the condition is True.
```

```python
# Example of sorting a Pandas DataFrame using Python:
# Example:

import pandas as pd

# Creating a sample DataFrame
data = {'Name': ['Ajay', 'Bhuvan', 'Chandhru', 'DK'],
        'Age': [25, 30, 22, 35],
        'Score': [85, 90, 78, 95]
}

df = pd.DataFrame(data)

# Sorting by 'Age' in ascending order
df_sorted_by_age = df.sort_values(by='Age')
print("Sorted by Age (Ascending):")
print(df_sorted_by_age)

# Sorting by 'Score' in descending order
df_sorted_by_score_desc = df.sort_values(by='Score', ascending=False)
print("\nSorted by Score (Descending):")
print(df_sorted_by_score_desc)

# Sorting by multiple columns (first by 'Score' descending, then by 'Age' ascending)
df_sorted_multiple = df.sort_values(by=['Score', 'Age'], ascending=[False, True])
print("\nSorted by Score (Descending) and then Age (Ascending):")
print(df_sorted_multiple)
```

# 1. Handling Missing Values:
    
# Identifying Missing Values

    Missing values can occur in any column of the DataFrame. 
    You can identify missing values with isna() or isnull(), 
    which return a DataFrame of boolean values indicating where the data is missing.

```python
import pandas as pd

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve'],
    'Age': [24, np.nan, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', None]
}

df = pd.DataFrame(data)

# Identify missing values
print(df.isna())
print(df.isnull())
```

```python
# In a pandas DataFrame, np.nan refers to "Not a Number" (NaN) from the NumPy library. It represents missing or undefined numerical data.

# Key Points About np.nan:
#     Acts as a placeholder for missing values in numerical columns.
#     Propagation in calculations: Most operations involving np.nan will result in np.nan (e.g., 5 + np.nan → np.nan).
#     Not equal to itself (np.nan == np.nan is False), so use pd.isna() or pd.isnull() to check for NaN values.
#     Can be handled using methods like fillna(), dropna(), etc.

import pandas as pd
import numpy as np

# Creating a DataFrame with NaN values
df = pd.DataFrame({'A': [1, 2, np.nan], 'B': [4, np.nan, 6]})

print(df)
```

# 3. Handling Outliers:
    
Identifying Outliers
Outliers are values that are significantly higher or lower than the majority of the data. You can identify outliers using methods like the Interquartile Range (IQR), which identifies the range within which most data points should lie.

```python
Using IQR to detect outliers:
```

```python
# Sample data with an outlier
data = {'Value': [10, 12, 15, 14, 100, 18, 20, 22, 24, 25]}
df = pd.DataFrame(data)

# Calculate Q1 and Q3
Q1 = df['Value'].quantile(0.25)
Q3 = df['Value'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Define the outlier range
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['Value'] < lower_bound) | (df['Value'] > upper_bound)]
print(outliers)
```

# Example: Full Data Cleaning Workflow

```python
import pandas as pd

# Sample DataFrame with missing values, duplicates, and outliers
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None, 'Eve', 'Bob', 'Alice'],
    'Age': [24, 27, 22, 32, 29, None, 24],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', None, 'Los Angeles', 'New York'],
    'Salary': [50000, 60000, 55000, 80000, 75000, 60000, 50000]
}

df = pd.DataFrame(data)

# Step 1: Handle Missing Values
df['Age'] = df['Age'].fillna(df['Age'].mean())  # Fill missing Age with the mean
df['City'] = df['City'].fillna('Unknown')  # Fill missing City with 'Unknown'

# Step 2: Remove Duplicates
df = df.drop_duplicates()

# Step 3: Identify and Handle Outliers (using Salary)
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['Salary'] = df['Salary'].clip(lower=lower_bound, upper=upper_bound)

# Step 4: Display cleaned data
print(df)
```

    Handling outliers: Use the IQR method to identify outliers and remove or cap them to improve data quality.
    Data cleaning ensures that your data is consistent, reliable, and ready for analysis or modeling.

# 1. loc and iloc Indexing
loc is label-based indexing, which means you use labels or index names to access data.
iloc is integer-based indexing, which means you use integer positions to access data (i.e., row/column positions as numbers).

```python
In pandas, .loc[] and .iloc[] are used for data selection and slicing, but they differ in how they reference rows and columns.

1. .loc[] → Label-based indexing:
                                                                     
	•	Uses explicit labels (row or column names).
	•	Can accept boolean arrays, lists of labels, or slicing with labels.
	•	Includes the end index in slicing.
```

```python
import pandas as pd

# Creating a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'Score': [85, 90, 95, 88]}

df = pd.DataFrame(data, index=['A', 'B', 'C', 'D'])

# Using .loc to select a single row
print(df.loc['B'])  # Select row with label 'B'

# Using .loc to select multiple rows
print(df.loc[['A', 'C']])  # Select rows 'A' and 'C'

# Using .loc for slicing (label-based, includes 'C')
print(df.loc['A':'C'])

# Selecting specific columns with .loc
print(df.loc[:, ['Name', 'Score']])  # Select all rows, but only 'Name' and 'Score' columns
```

```python
2. .iloc[] → Position-based indexing:

	•	Uses integer positions (like Python lists).
	•	End index in slicing is excluded.
	•	Accepts integer lists, slicing, or boolean arrays.
```

```python
# Using .iloc to select a single row by index position:

# Creating a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [25, 30, 35, 40],
        'Score': [85, 90, 95, 88]}

print(df.iloc[1])  # Select second row (index position 1)

# Using .iloc to select multiple rows by positions
print(df.iloc[[0, 2]])  # Select first and third rows

# Using .iloc for slicing (position-based, excludes index 2)
print(df.iloc[0:2])  # Select rows at positions 0 and 1

# Selecting specific columns with .iloc
print(df.iloc[:, [0, 2]])  # Select all rows, but only 'Name' and 'Score' columns
```

# groupby() for Grouping Data:
    
    -- The groupby() method in Pandas is a powerful tool to group data based on one or more columns. 
    -- It can be followed by aggregation functions such as sum(), mean(), or count() to summarize data.

# Example: Grouping by a Column

```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['AK', 'BK', 'CK', 'AK', 'BK', 'Ck'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'New York', 'Los Angeles'],
    'Age': [24, 27, 22, 32, 29, 28],
    'Salary': [50000, 60000, 55000, 80000, 75000, 72000]
}
df = pd.DataFrame(data)

print(df)

# Group by 'Name' and calculate the average salary for each person
Name_grouped = df.groupby('Name')['Salary'].mean()         # grouped = df.groupby('Name')['Salary'].sum()
print(Name_grouped)

# Group by 'Name' and calculate the average salary for each person
City_grouped = df.groupby('City')['Salary'].sum()         # grouped = df.groupby('City')['Salary'].sum()
print(City_grouped)
```

```python
# Aggregating Data with groupby()
You can apply multiple aggregation functions to grouped data using the agg() method.
```

```python
# Group by 'City' and get mean and sum of 'Age' and 'Salary'

print(df)

grouped_agg = df.groupby('City').agg({'Age': 'mean','Salary': 'sum'})
print(grouped_agg)
```

# Pivot Tables with pivot_table():
    
    The pivot_table() function is used to reshape data by creating a new table where you can specify rows,
    columns, and aggregation functions for data transformation.

# Example: Creating a Pivot Table

```python
# Exmaple - 1:
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice', 'Bob', 'Charlie'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Chicago', 'New York', 'Los Angeles'],
    'Age': [24, 27, 22, 32, 29, 28],
    'Salary': [50000, 60000, 55000, 80000, 75000, 72000]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Create a pivot table that shows the average Salary by City and Name

pivot = df.pivot_table(values='Salary', index='City', columns='Name', aggfunc='mean')
print(pivot)
```

# In the above o/p for pivot table, the rows represent City, the columns represent Name, and the values are the Salary with the mean function applied.

```python
# Example - 2:

import pandas as pd

# Sample data
data = {
    'Product': ['Apple', 'Banana', 'Carrot', 'Apple', 'Banana', 'Carrot', 'Apple', 'Carrot'],
    'Category': ['Fruit', 'Fruit', 'Vegetable', 'Fruit', 'Fruit', 'Vegetable', 'Fruit', 'Vegetable'],
    'Quantity Sold': [30, 45, 20, 50, 60, 35, 40, 55],
    'Revenue': [60, 90, 40, 100, 120, 70, 80, 110]
}

# Creating a DataFrame
df = pd.DataFrame(data)

print(df)
```

```python
# Pivot table for total quantity sold and revenue by category
pivot_table = pd.pivot_table(df, 
                             values=['Quantity Sold', 'Revenue'],             # Specifies which columns should be aggregated.
                             index='Category',                                # index defines the grouping column.
                             aggfunc={'Quantity Sold': 'sum', 'Revenue': 'sum'})  # aggfunc defines how values should be aggregated.

print(pivot_table)
```

```python
pivot_table = pd.pivot_table(df, values='Quantity Sold', index='Category', aggfunc='mean')
print(pivot_table)
```

```python
pivot_table = pd.pivot_table(df, values='Revenue', index='Category', aggfunc='count')
print(pivot_table)
```

```python
pivot_table = pd.pivot_table(df, values='Quantity Sold', index='Category', aggfunc='min')
print(pivot_table)
```

```python
pivot_table = pd.pivot_table(df, values='Quantity Sold', index='Category', aggfunc='max')
print(pivot_table)
```

```python
# Multiple Functions:
pivot_table = pd.pivot_table(df, values='Revenue', index='Category', aggfunc=['sum', 'mean'])
print(pivot_table)
```

```python
# Using Different Aggregations for Different Columns:
pivot_table = pd.pivot_table(df, 
                             values=['Quantity Sold', 'Revenue'], 
                             index='Category', 
                             aggfunc={'Quantity Sold': 'sum', 'Revenue': 'mean'})
print(pivot_table)
```

```python
# Example with Multiple Aggregation Functions:
pivot_table = pd.pivot_table(df, 
                             values=['Quantity Sold', 'Revenue'], 
                             index='Category', 
                             aggfunc={'Quantity Sold': ['sum', 'mean'], 'Revenue': ['sum', 'max']})
print(pivot_table)
```

```python
Example to see how df['Age'] (Series) and df[['Age']] (DataFrame) behave differently.

# Example DataFrame:

import pandas as pd
# Sample DataFrame
data = {'Name': ['AK', 'BK', 'CK'],
        'Age': [25, 30, 35],
        'Salary': [50000, 60000, 70000]}
df = pd.DataFrame(data)
print(df)

# 1. Using df['Age'] (Single Bracket - Series)

print(type(df['Age']))  # Output: <class 'pandas.core.series.Series'>

# Applying sum on Series
result = df['Age'].apply(sum)
print(result)

What happens here?
- `df['Age']` is a Series, so `.apply(sum)` tries to apply `sum` to each element.
- But `sum` is not meant for element-wise operations, so it just returns the same values.

# Output:

0    25
1    30
2    35
Name: Age, dtype: int64

This is incorrect because we wanted the sum of the entire column.

# 2. Using `df[['Age']]` (Double Brackets - DataFrame)

print(type(df[['Age']]))  # Output: <class 'pandas.core.frame.DataFrame'>

# Applying sum on DataFrame
result = df[['Age']].apply(sum)
print(result)

What happens here?
- `df[['Age']]` returns a DataFrame (not a Series).
- `.apply(sum)` now works column-wise and computes the sum of the `Age` column.

Output:

Age    90
dtype: int64

This is correct because it sums all the values in the column.

# Key Takeaways
| Selection Method | Data Type | Behavior |
|----------------- |----------- |---------- |
| `df['Age']`      | Series (`pandas.Series`)       | Element-wise operation, not summing the column |
| `df[['Age']]`    | DataFrame (`pandas.DataFrame`) | Column-wise operation, correctly sums the values |
```

# 3. Merging and Concatenating DataFrames

Merging DataFrames with merge()
Pandas provides a flexible and powerful merge() function to combine two DataFrames based on a common column (similar to SQL joins).

# Example: Merging DataFrames

```python
# Create two DataFrames with a common column 'Name'

df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [24, 27, 22]
})

df2 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Salary': [50000, 60000, 55000]
})

# Merge DataFrames on 'Name'
merged_df = pd.merge(df1, df2, on='Name')
print(merged_df)
```

# The merge() function performs an inner join by default,  meaning it combines only the rows with matching values in the Name column.

# Example: Left Join and Right Join Below:

```python
# Left join (all rows from df1(left), matching rows from df2(right))
left_joined = pd.merge(df1, df2, on='Name', how='left')
print(left_joined)
```

```python
# Right join (all rows from df2(right), matching rows from df1(left))
right_joined = pd.merge(df1, df2, on='Name', how='right')
print(right_joined)
```

# re.sub() in Python — Practical Examples

re.sub() is used to replace occurrences of a regex pattern with a specified replacement string.

🔹 Syntax: re.sub(pattern, replacement, string, count=0, flags=0)

```python
# Remove Special Characters:  --> [^\w\s] removes everything except letters, digits, and whitespace.

import re

text = "Hello!!! Welcome @ Python #2026"
cleaned = re.sub(r'[^\w\s]', '', text)

print(cleaned)
```

```python
# Remove Numbers: --> \d+ matches one or more digits.

text = "Order ID: 45982, Amount: 3500"
result = re.sub(r'\d+', '', text)

print(result)
```

```python
# Replace Multiple Spaces with Single Space:  --> \s+ matches multiple whitespace characters.

text = "Python     is    powerful"
result = re.sub(r'\s+', ' ', text)

print(result)
```

```python
# Remove URLs: --> http\S+ removes URL starting with http.

text = "Visit our site at https://example.com for details"
result = re.sub(r'http\S+', '', text)

print(result)
```

```python
# Replace Digits with #

text = "My phone number is 9876543210"
result = re.sub(r'\d', '#', text)

print(result)
```

```python
# Case-Insensitive Replacement:

text = "Python is python and PYTHON"
result = re.sub(r'python', 'Java', text, flags=re.IGNORECASE)

print(result)
```

```python
# Limit Number of Replacements:

text = "apple apple apple"
result = re.sub(r'apple', 'orange', text, count=2)

print(result)
```

```python
# Keep Only Alphabets:

text = "Python3.10 is #1!"
result = re.sub(r'[^a-zA-Z]', '', text)

print(result)
```

```python
# Common re.sub() Use Cases:

Use Case	             Pattern
Remove digits	         \d+
Remove special chars	 [^\w\s]
Remove spaces	         \s+
Remove URL	             http\S+
Remove emails	         \S+@\S+
Keep only letters	     [^a-zA-Z]
```

```python
# Use a Function as Replacement:

text = "The price is 50 and tax is 5"

def double(match):
    return str(int(match.group()) * 2)

result = re.sub(r'\d+', double, text)

print(result)

# When using a function:
# match.group() gives matched value
# Function returns replacement dynamically
```

```python
# Sample Dataset:

reviews = [
    "Amazing Product!!! 😍😍 Visit http://example.com",
    "Worst experience ever... 0/10",
    "Fast delivery @ 5pm!!!",
    "Quality is GOOD!!! #recommended",
]
```

```python
# Cleaning Requirements

Inside the function:
Convert text to lowercase.
Remove URLs.
Remove special characters.
Remove numbers.
Remove extra spaces.
Return the cleaned words as a list.
```

```python
import re

class ReviewPreprocessor:

    def preprocess_text(self, text: str) -> list:
        # Convert to lowercase
        text = text.lower()

        # Remove URLs using re.sub()
        text = re.sub(r'http\S+', '', text)

        # Remove numbers using re.sub()
        text = re.sub(r'', '', text)

        # Remove special characters using re.sub()
        text = re.sub(r'', '', text)

        # Remove extra spaces
        text = re.sub(r'\s+', ' ', text).strip()

        words = text.split()
        return words
```
