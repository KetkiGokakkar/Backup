````python
# Define the class
class Employee:
    # Constructor method
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    # Method to display employee details
    def display_details(self):
        print(f"ID: {self.emp_id}")
        print(f"Name: {self.name}")
        print(f"Salary: ₹{self.salary}")

    # Method to apply bonus
    def apply_bonus(self, percentage):
        bonus = self.salary * (percentage / 100)
        self.salary += bonus
        print(f"Bonus of {percentage}% applied. New salary: ₹{self.salary}")

# Create objects of the Employee class
emp1 = Employee(101, "Arjun", 50000)
emp2 = Employee(102, "Meera", 60000)

# Call methods on the objects
print("Employee 1 Details:")
emp1.display_details()
emp1.apply_bonus(10)

print("\nEmployee 2 Details:")
emp2.display_details()
emp2.apply_bonus(15)
````

````python
# Step 7: Using Data in Classes
# Let's create a class that handles more complex data. For example, a Car class with maintenance records:

class Car:
    def __init__(self, make, model, year, mileage, fuel_level):  # to initialize a car's basic information. 
        self.make = make                                         # self refers to the current object being created.
        self.model = model          # These lines assign the values passed to the constructor to the object’s instance variables using self.
        self.year = year
        self.mileage = mileage
        self.fuel_level = fuel_level
        
        self.maintenance_records = []  # Initializes an empty list to store maintenance records for the car.

    def drive(self, miles):
        self.mileage += miles
        self.fuel_level -= miles * 0.05 

    def refuel(self, gallons):
        self.fuel_level += gallons

    def add_maintenance(self, date, description):
        self.maintenance_records.append({"date": date, "description": description})
        # Adds a dictionary with the date and description of the maintenance to the maintenance_records list.
        
    def get_info(self):
        return f"{self.year} {self.make} {self.model} with {self.mileage} miles and {self.fuel_level} gallons of fuel"

    def get_maintenance_records(self):
        return self.maintenance_records  # Returns the list of all maintenance records (each as a dictionary).
````

````python
# Business Use Case: Online Payment Processing System:

Abstraction is the concept of hiding the internal implementation and showing only the necessary features. 
In Python, it’s typically done using abstract classes and methods via the abc module (Abstract Base Classes).
                                                                                                                                                                                                                                                                                                                                                                                          
A company supports different payment methods (like credit card, UPI, net banking), but the internal implementation for each is different. 
We want to define a common interface (abstract class) for all payment methods so the main system doesn’t worry about implementation details.
````

````python
from abc import ABC, abstractmethod

# Abstract class
class PaymentMethod(ABC):

    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def pay(self, amount):
        pass

# Concrete class 1
class CreditCardPayment(PaymentMethod):
    def authenticate(self):
        print("Authenticating Credit Card...")

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")

# Concrete class 2
class UpiPayment(PaymentMethod):
    def authenticate(self):
        print("Authenticating UPI...")

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")

# Concrete class 3
class NetBankingPayment(PaymentMethod):
    def authenticate(self):
        print("Authenticating Net Banking...")

    def pay(self, amount):
        print(f"Paid ₹{amount} using Net Banking")

# Business logic using abstraction
def process_payment(payment_method: PaymentMethod, amount: float):
    payment_method.authenticate()
    payment_method.pay(amount)

# Test the business logic
process_payment(CreditCardPayment(), 2500)
# process_payment(UpiPayment(), 1500)
# process_payment(NetBankingPayment(), 3000)
````

````python
# Key Benefits:

The main process_payment() function uses abstraction: it doesn’t care how payment happens.
Easy to add new payment methods without changing core business logic.
Promotes scalability and loose coupling.
````

````python
# D1 - Python - OOps - Abstraction:

from abc import ABC, abstractmethod

class Car(ABC):
    @abstractmethod
    def moveForward(self):
        pass

    @abstractmethod
    def moveBackward(self):
        pass

    @abstractmethod
    def fmRadio(self):
        pass

class Swift(Car):
    def __init__(self):
        self.__engineState = "off"  # Private attribute, hides internal state

    def __startEngine(self):        # Private method, hides internal logic
        self.__engineState = "on"
        print("Engine started")

    def __stopEngine(self):  # Private method
        self.__engineState = "off"
        print("Engine stopped")

    def moveForward(self):
        self.__startEngine()  # Using private method to handle internal details
        print("Swift is moving forward")
        self.__stopEngine()

    def moveBackward(self):
        self.__startEngine()  # Using private method to handle internal details
        print("Swift is moving backward")
        self.__stopEngine()

    def fmRadio(self):
        print("Swift FM radio is playing music")

# Create an instance of Swift and call its methods
swift = Swift()
swift.moveForward()
swift.moveBackward()
swift.fmRadio()
````

````python
# Encapsulation:

- Encapsulation can be achieved by declaring data members and methods of the class as **private**, which can be done with the help of Access modifers
- Access modifers makes the data/method public, private or protected
    1. Public Member: accesible anywhere using object reference
    2. Private Member: accessible within the current class
    3. Protected Member: Accessible within the class and its sub-classes
- But in Python, we don’t have direct access modifiers like public, private, and protected. We can achieve this by using single underscore and double underscores.
    1. Public Member: Normal
    2. Private Member: starts with __
    3. Protected Member: starts with _
````

````python
Let’s go **line-by-line** through your Python code that demonstrates **encapsulation**, and explain each part clearly:

---

### 🔐 **Encapsulation Concept**

Encapsulation means **hiding internal data** (by making it private) and **controlling access** via public methods (getters/setters).

---

### ✅ Code & Explanation

```python
class BankAccount:
    __account_holder = None   # Class-level private attribute
    __balance = 0             # Class-level private attribute
```

* `__account_holder` and `__balance` are **private** because of the double underscore `__`.
* These are **class variables**, shared across all instances unless shadowed.
* Note: Class variables aren't ideal for storing **instance-specific** data (we'll discuss this at the end).

---

```python
    def set_account_holder(self, name):
        self.__account_holder = name
```

* This is a **setter method**.
* It sets the private variable `__account_holder` using `self.__account_holder`.
* Here, `self.__account_holder` refers to an **instance-level** variable, which **shadows** the class-level one.

---

```python
    def get_account_holder(self):
        return self.__account_holder
```

* This is a **getter method**.
* It returns the value of the private `__account_holder` variable (at the instance level).

---

```python
    def get_balance(self):
        return self.__balance
```

* Another getter, this time for the balance.
* Also accessing the **instance-level** `__balance`.

---

```python
account = BankAccount()
```

* Creating an instance of the `BankAccount` class.

---

```python
account.set_account_holder("Sagar")
```

* This sets the private attribute `__account_holder` to `"Sagar"` using the **setter** method.

---

```python
# print(BankAccount.__account_holder)
# print(account.__account_holder)
```

* ❌ These two lines are **commented**.
* Even if you uncommented them, they would raise an **AttributeError**:

  * Python performs *name mangling* for private variables.
  * `__account_holder` becomes `_BankAccount__account_holder` internally.

---

```python
print(account._BankAccount__account_holder)
```

* ✅ This **bypasses** encapsulation by directly accessing the **name-mangled** variable.
* Not recommended, but possible: this accesses the private `__account_holder`.

---

```python
print(account.get_account_holder())
```

* ✅ This is the **correct and safe** way to access a private variable.
* Uses the **getter method** to read the value: `"Sagar"`.

---

### ⚠️ Bonus Note: Class vs Instance Variables

In your class:

```python
__account_holder = None
__balance = 0
```

These are **class-level** variables.

But inside the methods you use:

```python
self.__account_holder
```

This creates **instance-level** variables with name-mangling.

👉 To make the design clean, define them in the constructor (`__init__`):

```python
class BankAccount:
    def __init__(self):
        self.__account_holder = None
        self.__balance = 0
```

This ensures that **each object** has its **own copy** of these variables.
````

````python
# Encapsulation in Python – Proper Business Use Case:
Encapsulation means binding the data (variables) and the methods (functions) that operate on the data into a single unit, typically a class. 
It also restricts direct access to some of the object’s components, which is useful for protecting sensitive data.
````

````python
# Business Use Case: Bank Account Management System:

A bank wants to create a BankAccount class. The account balance and account holder details should be private and only accessible via proper methods 
(getters and setters). 
Direct access to internal variables must be restricted to prevent unauthorized changes.
````

````python
class BankAccount:
    def __init__(self, holder_name, balance):
        self.__account_holder = holder_name   # Private variable
        self.__balance = balance              # Private variable

    # Getter method for account holder
    def get_account_holder(self):
        return self.__account_holder

    # Setter method for account holder
    def set_account_holder(self, name):
        if name:  # basic validation
            self.__account_holder = name

    # Getter method for balance
    def get_balance(self):
        return self.__balance

    # Method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Invalid deposit amount.")

    # Method to withdraw money
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")
        else:
            print("Invalid or insufficient funds.")

# --- Test the Business Logic ---
account = BankAccount("Dinesh", 5000)

print("Account Holder:", account.get_account_holder())
print("Initial Balance: ₹", account.get_balance())

account.deposit(2000)
account.withdraw(1500)
print("Final Balance: ₹", account.get_balance())

# Trying to access private variables directly (Not recommended and will raise error)
# print(account.__balance)               # ❌ AttributeError
# print(account.__account_holder)        # ❌ AttributeError
````

````python
# Bank Account Example-2 - Best way to see public vs protected vs private in action.

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner                   # Public → can be accessed anywhere
        self._account_type = "Savings"       # Protected → internal use, not meant for direct external change
        self.__balance = balance             # Private → cannot be accessed directly
    
    # Public method (getter for private balance)
    def get_balance(self):
        return self.__balance
    
    # Public method (setter with validation)
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}, New Balance: {self.__balance}")
        else:
            print("Deposit must be positive")
    
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}, New Balance: {self.__balance}")
        else:
            print("Invalid withdrawal amount")

# --- Object Creation: ---
account = BankAccount("DK", 1000)

# Public (can access directly)
print("Owner:", account.owner)

# Protected (can access, but should avoid direct change)
print("Account Type (Protected):", account._account_type)

# Private (cannot access directly)
# print(account.__balance)   # Error
print("Balance via getter:", account.get_balance())   # Correct way

# Update balance using public methods
account.deposit(500)   # Deposited 500
account.withdraw(300)  # Withdrew 300

# Accessing private via name mangling (not recommended)
print("Hacked Balance:", account._BankAccount__balance)
````

````python
# Why It’s a Good Business Example of Encapsulation:

    Prevents unauthorized access to sensitive data (__balance).
    Only allowed operations (deposit, withdraw) can change the balance.
    Ensures control, validation, and business rules are enforced.
    Enhances data protection, security, and code reliability.
````

````python
"""
    Normal Methods: 
    Normal methods, or instance methods, are the most common type of methods in a class. 
    These methods can access and modify the object’s attributes. 
    They must have self as their first parameter, which is a reference to the instance calling the method.
"""
````

````python
"""
    Class Methods:
    Class methods are methods that are bound to the class itself, rather than to its instances. 
    They can modify the class state that applies across all instances of the class. 
    Class methods take cls as their first parameter, which is a reference to the class itself. 
    They are defined using the @classmethod decorator.
"""
````

````python
"""
    Class Variables:
        Class variables are data members that are shared across all instances of a class. 
        They are defined within the class but outside any instance methods. 
        All objects of the class share the same class variables,
        meaning any changes made to class variables will be reflected in all instances.
"""
````

````python
"""
    Static Methods:
    Static methods are methods that do not operate on an instance or class directly. 
    They do not require a reference to self or cls. 
    Static methods are defined using the @staticmethod decorator and are often used for utility functions.
"""
````

````python
"""
    Instance Variables:
        Instance variables are data members that are unique to each instance (object) of a class. 
        They are defined within the class and are typically initialized in the constructor. 
        Each object created from the class has its own copy of instance variables.
"""
````

````python
# Example -1 - Program for Class Variable - Instance Method and Variable:
class Student:    
    track="data"           # Class Variable (shared by all instances)
    
    def __init__(self,x,y):
        self.name=x        # Instance Variable (unique to each object)
        self.rollno=y      # Instance Variable (unique to each object)
        
s1=Student("Dinesh",10)    # Creating first object
s2=Student("Karthik",20)   # Creating second object 

print(s1.name,s1.rollno,s1.track)  # Output: Dinesh 10 data
print(s2.name,s2.rollno,s2.track)  # Output: Karthik 20 data

# Modifying class variable using an instance
s1.track = "AI"
print(s1.track)  # Output: AI (Modified only for s1)
print(s2.track)  # Output: data (Unaffected)
print(Student.track)  # Output: data (Class variable remains unchanged)

# Modifying class variable at the class level
Student.track = "Machine Learning"

print(s1.track)       # Output: AI (Still holds its modified value)  -- remains "AI", since it was modified at the instance level earlier.
print(s2.track)       # Output: Machine Learning (Updated from the class)
print(Student.track)  # Output: Machine Learning (Class variable updated)
````

````python
# Example -2 - Program for Class Variable - Instance Method / Class Method and Static Method
````

````python
class employee :
    company = "mindtree"   # Class Variable 
    
    # Defined inside the class but outside any method.Shared among all instances of the class.
    # Can be modified at the class level using a class method
    
    # Constructor: This method is by default called when a new instance is created -- self here takes the instance
    def __init__(self,first_name,last_name):  # init Method - __init__ is a constructor that initializes instance-specific 
                                              # attributes when a new object is created.
        self.first_name = first_name          # self.first_name and self.last_name - instance variables, unique for each object.
        self.last_name = last_name            # self refers to the instance of the class
        
    def generate_email(self):                       # instance Method -- operate on instance variables.
        return f'{self.first_name}.{self.last_name}@{employee.company}.com'  # This method generates an email using the class variable company.
    
    @classmethod                    # Class Method  Uses @classmethod decorator.cls refers to the class itself, not an instance. 
    def modify_company(cls,value):  # This method modifies the class variable company for all instances.
        cls.company = value
        
    @staticmethod                        # Static Method  # Does not use self or cls since it does not access instance or class attributes.
    def salary(stipend,monthy_sal):      # Works like a regular function inside a class.
        sal = stipend + monthy_sal       # Takes stipend and monthy_sal as parameters and returns the total salary.
        return sal
        
emp1 = employee('sai','rakesh')
emp2 = employee('ms','dhoni')

print(emp1.generate_email())
print(emp2.generate_email())

employee.modify_company('ltimindtree')
print(employee.company)

print(emp1.generate_email())
print(employee.salary(20000,30000))
````

````python
#Feature	         Definition	                           Example in Code
Class Variable	     Shared among all instances	           company = "mindtree"
Instance Variable	 Unique to each instance	           self.first_name = first_name
Instance Method	     Works on instance attributes	       generate_email(self)
Class Method	     Modifies class attributes	           modify_company(cls, value)
Static Method	     Regular function inside a class	   salary(stipend, monthy_sal)
````
